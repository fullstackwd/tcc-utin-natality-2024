### Escopo do pipeline analítico

| Etapa | Operação | Artefato de saída |
|:---:|---|---|
| 1 | Configuração do ambiente, registro de versões e organização dos diretórios | Metadados do pipeline e estrutura de pastas |
| 2 | Carga do arquivo `natality_data_2024.parquet` produzido pelo Notebook 01 | Base carregada para análise |
| 3 | Construção de variáveis derivadas, desfecho `NICU`, prematuridade, baixo peso, gemelaridade e proxies operacionais | Base analítica derivada |
| 4 | Validação metodológica das variáveis e documentação dos limites interpretativos da proxy RPM/PPROM | Checagens e contratos metodológicos |
| 5 | Geração de tabelas descritivas, completude, análise exploratória e regressões logísticas | Tabelas CSV do TCC |
| 6 | Produção dos gráficos exploratórios e gráficos de avaliação dos modelos | Imagens PNG dos gráficos |
| 7 | Treinamento e avaliação de modelos supervisionados nos cenários completo/perinatal e restritivo | Métricas, curvas ROC, PR, calibração e importância de variáveis |
| 8 | Geração de figuras metodológicas, quadros documentais e computacionais | Figuras e quadros do relatório |
| 9 | Conferência entre manifesto do TCC e arquivos efetivamente gerados | `manifesto_tcc_item_por_item.csv` |
| 10 | Exportação consolidada dos resultados e checklist final de reprodutibilidade | `resultados_tcc_pipeline_completo.xlsx` e arquivos auxiliares |

## Orientações de execução

<div style="margin: 18px 0; padding: 14px 18px; border-left: 4px solid #d97706; background: #fffbeb; border-radius: 4px; font-size: 13.5px; line-height: 1.6;">
  <strong style="color: #92400e; font-size: 13px; text-transform: uppercase; letter-spacing: 0.5px;">Execução sequencial obrigatória</strong>
  <p style="margin: 6px 0 0 0; color: #374151;">
    As células deste notebook devem ser executadas na ordem apresentada. Como o pipeline cria variáveis derivadas, diretórios, tabelas, gráficos, modelos e arquivos de auditoria de forma encadeada, a omissão ou execução fora de ordem de etapas intermediárias pode resultar em artefatos incompletos, métricas inconsistentes ou falhas na validação final de reprodutibilidade.
  </p>
</div>

**Pré-requisitos do ambiente:**

- Python 3.10 ou superior com kernel IPython
- Arquivo de entrada `natality_data_2024.parquet` previamente gerado pelo notebook 01, preferencialmente em `data/processed/`, ou informado por `NATALITY_PARQUET_PATH`
- Pacotes principais: `pandas`, `numpy`, `matplotlib`, `scikit-learn`, `statsmodels`, `xgboost` e `graphviz`
- Espaço em disco suficiente para geração de tabelas CSV, gráficos PNG, planilha consolidada XLSX, manifesto e arquivos auxiliares
- Memória RAM compatível com o processamento da base completa de natalidade
- Para execução final do TCC: não limitar artificialmente a amostra de análise
- Opcional: variável de ambiente `TCC_RF_CV_SAMPLE_MAX` para restringir a amostra em validações computacionais exploratórias
- Opcional: variável de ambiente `TCC_XGBOOST_DEVICE` para definir o dispositivo usado pelo XGBoost, quando aplicável

## Padrão enxuto de documentação das células

Cada etapa executável deve informar apenas o necessário para auditoria, evitando redundância com a introdução metodológica do notebook.

| Item | O que documentar | O que evitar |
|---|---|---|
| **Objetivo técnico** | A ação concreta executada pela célula. | Justificativas teóricas longas ou repetidas. |
| **Entrada** | Arquivos, parâmetros, variáveis, objetos ou artefatos consumidos. | Descrever novamente todo o estudo. |
| **Saída** | Tabela, gráfico, modelo, métrica, validação, arquivo ou estado computacional produzido. | Prometer artefatos que a célula não gera. |
| **Decisão metodológica** | Regra operacional que altera a interpretação, seleção de dados, definição de variável ou avaliação dos modelos. | Confundir configuração de ambiente com decisão analítica. |

Esta versão prioriza clareza, rastreabilidade e auditabilidade pela banca. A documentação deve permitir reconstruir a relação entre **código executado**, **artefato produzido** e **decisão metodológica aplicada**, sem transformar cada célula em uma revisão teórica.


## Matriz de aderência normativa e boas práticas aplicáveis

A tabela abaixo registra normas, princípios e guias usados como **referência metodológica e documental** para organizar o notebook. A aplicação é proporcional ao escopo do artefato: pipeline analítico reprodutível em Health Analytics, com microdados públicos de natalidade, criação de variáveis derivadas, análise estatística, modelagem supervisionada, exportação de resultados e auditoria por manifesto.

| Referência | Tema | Aplicação prática no notebook |
|---|---|---|
| **ISO/IEC 25012:2008 — Data Quality Model** | Qualidade de dados estruturados | Orienta completude, consistência, validade de tipos, tratamento de ausências, documentação de variáveis e validação dos artefatos. |
| **ISO 8000 — Data Quality** | Governança e qualidade de dados | Inspira registro de origem, linhagem analítica, regras de transformação, responsabilidade dos artefatos e rastreabilidade dos resultados. |
| **ISO 8601-1:2019 — Date and time format** | Padronização temporal | Recomenda datas e timestamps não ambíguos em registros de execução, nomes de arquivos, metadados, auditorias e evidências computacionais. |
| **ISO/IEC/IEEE 26512:2018 — Information for users** | Informação técnica para usuários | Fundamenta objetivo, escopo, pré-requisitos, entradas, saídas, limitações, decisões metodológicas e instruções de reprodução. |
| **ISO/IEC/IEEE 29119 — Software testing** | Verificação e validação | Apoia checagens de existência de arquivos, validação de variáveis, conferência de artefatos, controle de amostras e checklist final. |
| **ISO/IEC/IEEE 12207:2017 — Software life cycle processes** | Organização do ciclo de vida do artefato | Orienta a sequência configuração → carga → transformação → análise → modelagem → validação → exportação → auditoria. |
| **ISO/IEC 25010:2011 — Software product quality model** | Qualidade do artefato computacional | Orienta manutenibilidade, confiabilidade, portabilidade, usabilidade e reprodutibilidade do notebook. |
| **ISO/IEC 27001 e ISO/IEC 27002** | Segurança da informação | Reforça cautela com acesso, armazenamento, compartilhamento e rastreabilidade de dados relacionados à saúde, mesmo quando públicos. |
| **LGPD — Lei nº 13.709/2018** | Proteção de dados pessoais | Reforça finalidade, necessidade, minimização, segurança, transparência e cautela ética no uso de microdados relacionados a nascimento e saúde. |
| **PEP 8** | Estilo de código Python | Orienta legibilidade, organização de imports, espaçamento, nomes claros e consistência visual das células executáveis. |

| **FAIR Principles** | Dados encontráveis, acessíveis, interoperáveis e reutilizáveis | Inspira metadados, nomenclatura clara, documentação de origem, preservação de artefatos e possibilidade de reuso dos resultados. |
| **DDI-Codebook / Data Documentation Initiative** | Documentação de microdados | Inspira descrição de variáveis, códigos, origem, estrutura do arquivo, critérios de transformação e limitações analíticas. |
| **Frictionless Data Package** | Metadados para dados tabulares | Inspira documentação de recursos, campos, tipos, formatos, arquivos de saída e relações entre tabelas geradas. |
| **STROBE** | Relato de estudos observacionais | Apoia documentação do desenho retrospectivo, fonte de dados, população, variáveis, vieses, limitações e interpretação associativa. |
| **TRIPOD / TRIPOD+AI** | Relato de modelos preditivos em saúde | Orienta transparência sobre variáveis, população, divisão treino-teste, métricas, calibração, uso de machine learning e limitações. |

### Critério de aplicação

As referências foram aplicadas como **boas práticas de engenharia de dados, ciência de dados em saúde, documentação científica e reprodutibilidade computacional**, com os seguintes focos:

1. **Rastreabilidade:** fonte dos dados, caminhos, parâmetros, variáveis, decisões operacionais, artefatos gerados e conferência por manifesto.
2. **Reprodutibilidade:** execução sequencial, dependências explícitas, configurações centralizadas, registro de ambiente, salvamento de tabelas, gráficos, métricas e checklist final.
3. **Qualidade de dados:** completude, consistência de tipos, tratamento de códigos especiais, validação de variáveis derivadas e conferência das saídas.
4. **Transparência metodológica:** documentação das decisões que afetam interpretação, especialmente o uso de proxy operacional compatível com contexto de RPM/PPROM, sem equivalência diagnóstica individual.
5. **Modelagem auditável:** separação entre análise estatística e modelagem supervisionada, registro de cenários, métricas, calibração, importância de variáveis e limitações.
6. **Segurança e ética:** minimização do escopo, uso de dados públicos, cautela com variáveis de saúde e interpretação restrita a achados observacionais e associativos.

> **Nota metodológica:** a presença dessas referências indica aderência documental, procedimental e computacional a boas práticas reconhecidas. Não implica certificação ISO, conformidade legal plena, validação clínica, auditoria regulatória, aprovação ética institucional ou aptidão para uso clínico prospectivo. Esses níveis de conformidade exigiriam processo formal próprio, evidências adicionais e avaliação independente.


## Glossário técnico do pipeline

| Termo | Significado no notebook |
|---|---|
| **Artefato analítico** | Arquivo, tabela, gráfico, modelo ou planilha produzido pelo pipeline e utilizado como evidência computacional do TCC. |
| **Base analítica** | Conjunto de dados estruturado após seleção, limpeza, transformação e criação de variáveis derivadas. |
| **Microdados públicos** | Registros individuais disponibilizados por órgão público, usados de forma retrospectiva e sem identificação direta no contexto deste notebook. |
| **Pipeline analítico** | Sequência organizada de etapas computacionais que vai da leitura da base à geração de resultados, modelos, gráficos, manifesto e checklist. |
| **Execução sequencial** | Regra segundo a qual as células devem ser executadas na ordem apresentada para preservar dependências entre objetos e artefatos. |
| **Variável derivada** | Variável criada a partir de uma ou mais variáveis originais, conforme regra operacional documentada no notebook. |
| **Proxy operacional** | Variável aproximada usada para representar um conceito clínico ou metodológico que não está diretamente disponível na base original. |
| **RPM/PPROM** | Contexto clínico relacionado à ruptura prematura de membranas; neste notebook, tratado apenas por aproximação operacional e com interpretação cautelosa. |
| **UTIN/NICU** | Unidade de Terapia Intensiva Neonatal; usada como desfecho ou variável de interesse conforme a etapa analítica. |
| **Prematuridade** | Condição associada ao nascimento antes do termo gestacional, operacionalizada no notebook a partir da idade gestacional disponível. |
| **Missing** | Valor ausente, inválido, desconhecido, não informado ou codificado como ignorado na base original. |
| **Completude** | Percentual de registros com informação válida para uma variável específica. |
| **Validação estrutural** | Checagem de existência, formato, colunas esperadas, tipos de dados e consistência básica dos artefatos produzidos. |
| **Regressão logística** | Modelo estatístico usado para estimar associação entre variáveis explicativas e um desfecho binário. |
| **Modelo supervisionado** | Algoritmo treinado com pares de entrada e desfecho conhecido para estimar probabilidades ou classificações. |
| **Calibração** | Avaliação da concordância entre probabilidades previstas pelo modelo e frequências observadas do desfecho. |
| **Curva ROC** | Gráfico que avalia a capacidade discriminatória do modelo em diferentes pontos de corte. |
| **Curva Precision-Recall** | Gráfico útil para avaliar desempenho preditivo em cenários com desfecho desbalanceado. |
| **Importância de variáveis** | Medida usada para indicar a contribuição relativa de cada variável no desempenho de modelos supervisionados. |
| **Manifesto do TCC** | Arquivo de conferência que compara os artefatos esperados com os arquivos efetivamente gerados. |
| **Checklist de reprodutibilidade** | Lista de verificação final usada para auditar se as etapas essenciais do pipeline foram executadas e documentadas. |
| **Auditoria de execução** | Registro técnico que permite reconstituir como, quando, com quais parâmetros e com quais saídas o pipeline foi executado. |

````python
# 0. Instalação e verificação de dependências obrigatórias
# -------------------------------------------------------------------------
# Finalidade:
# - preparar o ambiente Python/Jupyter antes da execução do pipeline;
# - verificar se as bibliotecas obrigatórias estão disponíveis;
# - instalar automaticamente pacotes ausentes usando o mesmo Python do kernel;
# - tratar dependência opcional de extração Deflate64 sem interromper o notebook;
# - emitir mensagens claras para apoiar auditoria e reprodutibilidade.
#
# Entrada:
# - ambiente Python/Jupyter ativo;
# - acesso ao gerenciador de pacotes pip;
# - lista centralizada de pacotes obrigatórios e opcionais.
#
# Saída:
# - ambiente validado para as próximas etapas do notebook;
# - mensagens indicando dependências já disponíveis, instaladas ou ausentes;
# - reinício automático do kernel quando uma dependência obrigatória for instalada.
#
# Decisão metodológica:
# - esta célula tem finalidade operacional. Ela não altera a base de dados, as
#   variáveis analíticas, os critérios de inclusão, os modelos ou a interpretação
#   dos resultados do TCC.

from __future__ import annotations

import importlib
import subprocess
import sys
import textwrap
from typing import Dict


REQUIRED_PACKAGES: Dict[str, str] = {
    "numpy": "numpy",
    "pandas": "pandas",
    "requests": "requests",
    "pyarrow": "pyarrow",
}

OPTIONAL_PACKAGES: Dict[str, str] = {
    "zipfile_deflate64": "zipfile-deflate64",
}


def module_available(module_name: str) -> bool:
    """Verifica se um módulo Python está disponível para importação.

    Parameters
    ----------
    module_name : str
        Nome do módulo a ser testado.

    Returns
    -------
    bool
        True se o módulo puder ser importado; False caso contrário.
    """
    try:
        importlib.import_module(module_name)
        return True
    except Exception:
        return False


def install_package(package_name: str, *, required: bool = True) -> bool:
    """Instala um pacote via pip usando o mesmo Python do kernel.

    Parameters
    ----------
    package_name : str
        Nome do pacote no pip.
    required : bool, default=True
        Indica se a falha de instalação deve interromper o notebook.

    Returns
    -------
    bool
        True se a instalação foi concluída; False se era opcional e falhou.

    Raises
    ------
    RuntimeError
        Se uma dependência obrigatória não puder ser instalada.
    """
    command = [
        sys.executable,
        "-m",
        "pip",
        "install",
        "--upgrade",
        "--prefer-binary",
        package_name,
    ]

    print(f"Dependência ausente. Instalando: {' '.join(command)}")

    try:
        subprocess.check_call(command)
        return True

    except subprocess.CalledProcessError as exc:
        if required:
            manual_command = f"{sys.executable} -m pip install --upgrade --prefer-binary {package_name}"
            raise RuntimeError(
                textwrap.dedent(
                    f"""
                    Falha ao instalar dependência obrigatória: {package_name}

                    Execute manualmente no terminal:

                        {manual_command}

                    Depois reinicie o kernel e execute o notebook novamente.
                    """
                ).strip()
            ) from exc

        print(f"AVISO: dependência opcional não instalada: {package_name}")
        print("O notebook continuará usando métodos alternativos quando disponíveis.")
        return False


installed_required_packages: list[str] = []

for module_name, package_name in REQUIRED_PACKAGES.items():
    if module_available(module_name):
        print(f"ATIVIDADE JÁ REALIZADA: dependência obrigatória disponível: {module_name}")
        continue

    print(f"Dependência obrigatória ausente: {module_name} -> pacote {package_name}")
    if install_package(package_name, required=True):
        installed_required_packages.append(package_name)

for module_name, package_name in OPTIONAL_PACKAGES.items():
    if module_available(module_name):
        print(f"ATIVIDADE JÁ REALIZADA: dependência opcional disponível: {module_name}")
        continue

    print(f"Dependência opcional ausente: {module_name} -> pacote {package_name}")
    install_package(package_name, required=False)

if installed_required_packages:
    print("\nDependências obrigatórias instaladas:")
    for package_name in installed_required_packages:
        print(f"- {package_name}")

    print("\nReiniciando o kernel para ativar as dependências recém-instaladas...")
    try:
        from IPython import get_ipython

        ipython = get_ipython()
        if ipython is not None:
            ipython.kernel.do_shutdown(restart=True)
        else:
            print("Kernel IPython não identificado. Reinicie o ambiente manualmente.")

    except Exception:
        print("Não foi possível reiniciar automaticamente. Reinicie o kernel manualmente.")

else:
    print("\nATIVIDADE JÁ REALIZADA: dependências obrigatórias verificadas. Nenhuma instalação obrigatória foi necessária.")
````

          e
          <code style="background: #dbeafe; color: #1e3a8a; padding: 1px 4px; border-radius: 3px;">XGBOOST_VERSAO_MAJOR</code>.
          A célula também exibe uma tabela-resumo com a configuração efetivamente aplicada.
        </td>
      </tr>
      <tr>
        <td style="padding: 7px 9px; font-weight: 600; color: #1d4ed8;">Observação para o TCC</td>
        <td style="padding: 7px 9px; color: #1e40af;">
          Esta configuração aumenta a portabilidade do pipeline, pois permite execução acelerada em ambientes com GPU
          e mantém alternativa segura em CPU em ambientes acadêmicos sem CUDA. A decisão afeta principalmente o tempo
          de processamento dos modelos, não a definição das variáveis, o desenho analítico ou a interpretação
          metodológica do TCC. Para reprodutibilidade, o modo solicitado, a versão do XGBoost, a detecção de CUDA
          e a estratégia efetiva devem ser registrados junto aos resultados.
        </td>
      </tr>
    </tbody>
  </table>

</div>

````python
# Configuração robusta de aceleração do XGBoost
# Esta célula deve ser executada antes da modelagem. Ela também é chamada como fallback
# dentro da célula dos modelos para evitar NameError em execuções parciais.

def detectar_cuda_nvidia_smi():
    """Retorna True quando o comando nvidia-smi está disponível e detecta uma GPU CUDA."""
    try:
        resultado = subprocess.run(
            ["nvidia-smi"],
            stdout=subprocess.PIPE,
            stderr=subprocess.PIPE,
            text=True,
            timeout=5,
        )
        return resultado.returncode == 0
    except Exception:
        return False

def _versao_xgboost_major():
    if not XGBOOST_AVAILABLE or xgb is None:
        return None
    try:
        return int(str(xgb.__version__).split(".")[0])
    except Exception:
        return None

def configurar_aceleracao_xgboost():
    """
    Define parâmetros de aceleração para XGBoost com fallback seguro.

    Variável de ambiente:
    - TCC_XGBOOST_DEVICE=auto: usa GPU se CUDA for detectada; caso contrário, CPU.
    - TCC_XGBOOST_DEVICE=gpu: solicita GPU; se CUDA não for detectada, cai para CPU.
    - TCC_XGBOOST_DEVICE=cpu: força CPU.
    """
    modo = os.environ.get("TCC_XGBOOST_DEVICE", "auto").strip().lower()
    if modo not in {"auto", "gpu", "cuda", "cpu"}:
        modo = "auto"

    cuda_detectado = detectar_cuda_nvidia_smi()
    major = _versao_xgboost_major()

    solicitar_gpu = modo in {"auto", "gpu", "cuda"} and cuda_detectado and XGBOOST_AVAILABLE

    if solicitar_gpu and major is not None and major >= 2:
        parametros = {
            "tree_method": "hist",
            "device": "cuda",
        }
        usando_gpu = True
        estrategia = 'XGBoost 2.x: tree_method="hist", device="cuda"'
    elif solicitar_gpu:
        parametros = {
            "tree_method": "gpu_hist",
            "predictor": "gpu_predictor",
        }
        usando_gpu = True
        estrategia = 'XGBoost 1.x: tree_method="gpu_hist", predictor="gpu_predictor"'
    else:
        parametros = {
            "tree_method": "hist",
        }
        usando_gpu = False
        estrategia = 'CPU/fallback: tree_method="hist"'

    return {
        "MODO_ACELERACAO_XGBOOST": modo,
        "CUDA_DETECTADO": bool(cuda_detectado),
        "USAR_GPU_XGBOOST": bool(usando_gpu),
        "PARAMETROS_XGBOOST_ACELERACAO": parametros,
        "ESTRATEGIA_XGBOOST_ACELERACAO": estrategia,
        "XGBOOST_VERSAO_MAJOR": major,
    }

_config_xgb_gpu = configurar_aceleracao_xgboost()
MODO_ACELERACAO_XGBOOST = _config_xgb_gpu["MODO_ACELERACAO_XGBOOST"]
CUDA_DETECTADO = _config_xgb_gpu["CUDA_DETECTADO"]
USAR_GPU_XGBOOST = _config_xgb_gpu["USAR_GPU_XGBOOST"]
PARAMETROS_XGBOOST_ACELERACAO = _config_xgb_gpu["PARAMETROS_XGBOOST_ACELERACAO"]
ESTRATEGIA_XGBOOST_ACELERACAO = _config_xgb_gpu["ESTRATEGIA_XGBOOST_ACELERACAO"]
XGBOOST_VERSAO_MAJOR = _config_xgb_gpu["XGBOOST_VERSAO_MAJOR"]

display(pd.DataFrame([
    {"Item": "Modo solicitado", "Valor": MODO_ACELERACAO_XGBOOST},
    {"Item": "CUDA detectado via nvidia-smi", "Valor": CUDA_DETECTADO},
    {"Item": "XGBoost disponível", "Valor": XGBOOST_AVAILABLE},
    {"Item": "Versão major do XGBoost", "Valor": XGBOOST_VERSAO_MAJOR},
    {"Item": "Usando GPU no XGBoost", "Valor": USAR_GPU_XGBOOST},
    {"Item": "Estratégia efetiva", "Valor": ESTRATEGIA_XGBOOST_ACELERACAO},
    {"Item": "Parâmetros aplicados", "Valor": json.dumps(PARAMETROS_XGBOOST_ACELERACAO, ensure_ascii=False)},
]))
````

````python
# ============================================================
# Carga revisada do artefato analítico gerado pelo notebook 01
# ============================================================
# Padrões aderentes: ISO/IEC 25012; ISO 8000; FAIR Principles; DDI-Codebook.
# Finalidade:
# - consumir diretamente o arquivo `natality_data_2024.parquet` produzido pelo notebook 01;
# - eliminar definitivamente a reconstrução do TXT fixed-width no notebook 02;
# - validar schema mínimo, domínios binários e tipos numéricos necessários às etapas seguintes;
# - registrar evidência auditável da carga do artefato de entrada;
# - manter compatibilidade com células posteriores que usam `df`, `data` e `COLSPECS.keys()`.
#
# Entrada obrigatória:
# - `natality_data_2024.parquet`, preferencialmente em `data/processed/`.
#
# Alternativa operacional:
# - definir a variável de ambiente NATALITY_PARQUET_PATH com o caminho completo do arquivo.
#
# Observação metodológica:
# - aquisição, extração, leitura fixed-width, limpeza inicial e auditoria estrutural pertencem ao notebook 01;
# - este notebook inicia a análise do TCC a partir do Parquet previamente preparado.

from pathlib import Path
import hashlib
import os

NOME_ARQUIVO_PARQUET_01 = "natality_data_2024.parquet"

# Layout documental preservado do notebook 01.
# O notebook 02 não usa essas posições para leitura fixed-width; elas são mantidas
# como dicionário técnico mínimo e para compatibilidade com células que usam COLSPECS.keys().
COLSPECS = {
    "MAGER": (74, 76),
    "PREVIS": (237, 239),
    "WIC": (250, 251),
    "RF_GDIAB": (313, 314),
    "RF_GHYPE": (315, 316),
    "LD_INDL": (382, 383),
    "LD_STER": (384, 385),
    "LD_ANTB": (385, 386),
    "LD_CHOR": (386, 387),
    "APGAR5": (443, 445),
    "DPLURAL": (453, 454),
    "SETORDER_R": (458, 459),
    "COMBGEST": (489, 491),
    "GESTREC3": (493, 494),
    "DBWT": (503, 507),
    "BWTR4": (510, 511),
    "AB_AVEN1": (516, 517),
    "AB_AVEN6": (517, 518),
    "AB_NICU": (518, 519),
    "AB_SURF": (519, 520),
    "AB_ANTI": (520, 521),
    "AB_SEIZ": (521, 522),
}

VARIAVEIS_ORIGINAIS_NATALITY = list(COLSPECS.keys())

VARIAVEIS_DERIVADAS_NOTEBOOK01 = [
    "PRETERM",
    "LOW_BW",
    "VERY_LOW_BW",
    "LOW_APGAR5",
    "MULTIPLA",
    "GEMELAR",
    "PROM_PROXY_WIDE",
    "PROM_PROXY_CONSERV",
    "DESFECHO_NEONATAL_GRAVE",
]

COLUNAS_OBRIGATORIAS_PARQUET_01 = VARIAVEIS_ORIGINAIS_NATALITY + VARIAVEIS_DERIVADAS_NOTEBOOK01

COLUNAS_NUMERICAS_ESPERADAS = sorted(set(
    VARIAVEIS_ORIGINAIS_NATALITY +
    VARIAVEIS_DERIVADAS_NOTEBOOK01 +
    ["AB_NICU"]
))

COLUNAS_BINARIAS_ESPERADAS = [
    "WIC", "RF_GDIAB", "RF_GHYPE",
    "LD_INDL", "LD_STER", "LD_ANTB", "LD_CHOR",
    "AB_AVEN1", "AB_AVEN6", "AB_NICU", "AB_SURF", "AB_ANTI", "AB_SEIZ",
    "PRETERM", "LOW_BW", "VERY_LOW_BW", "LOW_APGAR5", "MULTIPLA", "GEMELAR",
    "PROM_PROXY_WIDE", "PROM_PROXY_CONSERV", "DESFECHO_NEONATAL_GRAVE",
]


def sha256_arquivo(caminho: Path, bloco_bytes: int = 1024 * 1024) -> str:
    """Calcula SHA-256 do arquivo de entrada para rastreabilidade.

    Norma técnica aplicada:
    - ISO 8000: rastreabilidade do artefato de dados.
    - FAIR Principles: identificação persistente e verificável do recurso usado.
    """
    digest = hashlib.sha256()
    with Path(caminho).open("rb") as f:
        for bloco in iter(lambda: f.read(bloco_bytes), b""):
            digest.update(bloco)
    return digest.hexdigest()


def _caminhos_candidatos_parquet(nome_arquivo: str = NOME_ARQUIVO_PARQUET_01) -> list[Path]:
    """Gera candidatos determinísticos para o Parquet do notebook 01.

    A função evita varreduras recursivas amplas em diretórios pessoais, pois esse
    padrão pode tornar a execução lenta em servidores, Colab, drives montados ou
    ambientes compartilhados.
    """
    candidatos: list[Path] = []

    env_path = os.environ.get("NATALITY_PARQUET_PATH")
    if env_path:
        candidatos.append(Path(env_path).expanduser())

    raizes = [BASE_DIR, Path.cwd(), Path("/mnt/data")]
    subpastas_preferidas = [
        Path("data/processed"),
        Path("dados_processados"),
        Path("dados_finais"),
        Path("dados_intermediarios"),
        Path("dados"),
        Path("data"),
        Path("outputs"),
    ]

    # Também considera pais próximos do diretório de execução. Isso cobre casos em
    # que o notebook 02 está em uma subpasta do projeto e o notebook 01 gerou o
    # Parquet em ../data/processed.
    for parent in list(Path.cwd().resolve().parents)[:3]:
        raizes.append(parent)

    vistos = set()
    for raiz in raizes:
        try:
            raiz_resolvida = raiz.expanduser().resolve()
        except Exception:
            raiz_resolvida = raiz
        if str(raiz_resolvida) in vistos:
            continue
        vistos.add(str(raiz_resolvida))

        candidatos.append(raiz_resolvida / nome_arquivo)
        for subpasta in subpastas_preferidas:
            candidatos.append(raiz_resolvida / subpasta / nome_arquivo)

    # Remove duplicatas preservando ordem de prioridade.
    candidatos_unicos = []
    vistos = set()
    for candidato in candidatos:
        chave = str(candidato)
        if chave not in vistos:
            vistos.add(chave)
            candidatos_unicos.append(candidato)

    return candidatos_unicos


def localizar_parquet_notebook01(nome_arquivo: str = NOME_ARQUIVO_PARQUET_01) -> Path | None:
    """Localiza o Parquet produzido pelo notebook 01 em caminhos previsíveis."""
    for candidato in _caminhos_candidatos_parquet(nome_arquivo):
        if candidato.exists() and candidato.is_file():
            return candidato
    return None


def listar_arquivos_relevantes(max_itens: int = 60) -> list[Path]:
    """Lista arquivos de dados próximos para diagnóstico quando o Parquet não é encontrado."""
    padroes = ["*.parquet", "*.PARQUET", "*.csv", "*.CSV", "*.txt", "*.TXT", "*.zip", "*.ZIP"]
    raizes = [BASE_DIR, Path.cwd(), Path("/mnt/data")]
    encontrados: list[Path] = []

    for raiz in raizes:
        try:
            raiz = raiz.expanduser().resolve()
        except Exception:
            continue
        if not raiz.exists():
            continue
        for padrao in padroes:
            encontrados.extend(raiz.glob(padrao))
            for subpasta in ["data", "data/processed", "dados", "dados_finais", "dados_intermediarios", "outputs"]:
                sub = raiz / subpasta
                if sub.exists():
                    encontrados.extend(sub.glob(padrao))

    unicos = []
    vistos = set()
    for caminho in encontrados:
        chave = str(caminho.resolve()) if caminho.exists() else str(caminho)
        if chave not in vistos:
            vistos.add(chave)
            unicos.append(caminho)

    return sorted(unicos, key=lambda p: (len(str(p)), str(p)))[:max_itens]


def validar_e_normalizar_schema_parquet_notebook01(base: pd.DataFrame) -> pd.DataFrame:
    """Valida e normaliza o schema mínimo esperado do artefato do notebook 01.

    Norma técnica aplicada:
    - ISO/IEC 25012: completude, consistência e conformidade estrutural.
    - ISO 8000: controle de qualidade e rastreabilidade do dado.
    - DDI-Codebook: preservação da documentação das variáveis do microdado.
    """
    base = base.copy()

    ausentes = [col for col in COLUNAS_OBRIGATORIAS_PARQUET_01 if col not in base.columns]
    if ausentes:
        raise ValueError(
            "O arquivo encontrado não possui o schema mínimo esperado do notebook 01.\n"
            f"Colunas ausentes ({len(ausentes)}): {ausentes}\n"
            "Reexecute o notebook 01 e confirme a geração de `natality_data_2024.parquet`."
        )

    # Coerção controlada para reduzir risco de erro em comparações numéricas nas células posteriores.
    for coluna in COLUNAS_NUMERICAS_ESPERADAS:
        if coluna in base.columns:
            base[coluna] = pd.to_numeric(base[coluna], errors="coerce")

    dominios_invalidos = {}
    for coluna in COLUNAS_BINARIAS_ESPERADAS:
        if coluna not in base.columns:
            continue
        valores = set(base[coluna].dropna().unique().tolist())
        valores_normalizados = set()
        for valor in valores:
            try:
                valores_normalizados.add(int(float(valor)))
            except Exception:
                valores_normalizados.add(valor)
        if not valores_normalizados.issubset({0, 1}):
            dominios_invalidos[coluna] = sorted(valores_normalizados, key=lambda x: str(x))[:20]

    if dominios_invalidos:
        raise ValueError(
            "Foram encontrados valores fora do domínio binário esperado {0, 1}.\n"
            f"Variáveis com problema: {dominios_invalidos}\n"
            "Verifique a etapa de preparo no notebook 01."
        )

    if base.empty:
        raise ValueError("O Parquet foi localizado, mas a base carregada está vazia.")

    return base


ARQUIVO_PARQUET_NOTEBOOK01 = localizar_parquet_notebook01()

if ARQUIVO_PARQUET_NOTEBOOK01 is None:
    arquivos_diag = listar_arquivos_relevantes()
    caminhos_verificados = _caminhos_candidatos_parquet()
    mensagem = [
        "Arquivo obrigatório não encontrado: natality_data_2024.parquet",
        "",
        "Este notebook 02 foi revisado para consumir o artefato produzido pelo notebook 01.",
        "Execute primeiro o notebook 01 e confirme a geração do Parquet em um dos caminhos esperados.",
        "",
        "Caminhos verificados:",
    ]
    mensagem.extend([f"  - {p}" for p in caminhos_verificados[:40]])
    mensagem.extend([
        "",
        "Alternativa: defina a variável de ambiente NATALITY_PARQUET_PATH com o caminho completo do arquivo.",
        "",
        "Arquivos de dados encontrados para diagnóstico:",
    ])
    if arquivos_diag:
        mensagem.extend([f"  - {p}" for p in arquivos_diag[:40]])
    else:
        mensagem.append("  - nenhum arquivo de dados encontrado nas pastas verificadas.")
    raise FileNotFoundError("\n".join(mensagem))

ARQUIVO_PARQUET = ARQUIVO_PARQUET_NOTEBOOK01

df = pd.read_parquet(ARQUIVO_PARQUET)
df = validar_e_normalizar_schema_parquet_notebook01(df)

METADADOS_BASE_CARREGADA = {
    "arquivo": str(ARQUIVO_PARQUET.resolve()),
    "nome_arquivo": ARQUIVO_PARQUET.name,
    "origem_pipeline": "notebook 01 - preparo reprodutível da Natality 2024",
    "linhas": int(df.shape[0]),
    "colunas": int(df.shape[1]),
    "colunas_obrigatorias_conferidas": int(len(COLUNAS_OBRIGATORIAS_PARQUET_01)),
    "tamanho_bytes": int(ARQUIVO_PARQUET.stat().st_size),
    "sha256": sha256_arquivo(ARQUIVO_PARQUET),
}

# Registro auditável da carga do artefato do notebook 01.
try:
    DIR_MANIFESTO.mkdir(parents=True, exist_ok=True)
    pd.DataFrame([METADADOS_BASE_CARREGADA]).to_csv(
        DIR_MANIFESTO / "carga_parquet_notebook01.csv",
        index=False,
        encoding="utf-8-sig",
    )
except Exception as exc:
    print("AVISO: não foi possível salvar auditoria da carga do Parquet:", exc)

print("Parquet do notebook 01 carregado:", ARQUIVO_PARQUET)
print("Dimensão da base carregada:", df.shape)
print("SHA-256 do arquivo de entrada:", METADADOS_BASE_CARREGADA["sha256"])

try:
    display(pd.DataFrame([METADADOS_BASE_CARREGADA]))
    display(df.head())
except Exception:
    print(METADADOS_BASE_CARREGADA)
````

## 4. Variáveis derivadas e proxies operacionais

Esta seção cria as variáveis analíticas usadas no estudo, incluindo o desfecho principal `NICU`, marcadores de prematuridade, baixo peso ao nascer, gemelaridade, grupos gestacionais e variáveis compostas.

As proxies compatíveis com contexto de RPM/PPROM são construídas apenas para análise retrospectiva. Elas combinam prematuridade com marcadores disponíveis no arquivo público, mas não substituem uma variável diagnóstica direta, não confirmam RPM/PPROM em nível individual e não estimam a prevalência clínica real da condição.

A proxy conservadora (`PROM_PROXY_CONSERV`) prioriza maior especificidade analítica ao exigir prematuridade e corioamnionite. A proxy ampla (`PROM_PROXY_WIDE`) prioriza maior sensibilidade analítica ao combinar prematuridade com corioamnionite, antibioticoterapia intraparto ou uso de esteroides. Como `LD_STER` e `LD_ANTB` também podem refletir diferentes contextos obstétricos e assistenciais associados à prematuridade, a proxy ampla deve ser interpretada como **marcador operacional ampliado de vulnerabilidade pré-termo**, e não como aproximação diagnóstica individual de RPM/PPROM.

Há ainda uma decisão operacional relevante: na proxy ampla, quando `PRETERM` está informado e nenhum dos três indicadores componentes está positivo, o registro é classificado como negativo para a proxy. Essa regra preserva a variável binária necessária às análises, mas pode subestimar registros potencialmente compatíveis quando houver ausência ou sub-registro dos componentes clínico-assistenciais.


<div style="margin: 14px 0 22px 0; padding: 14px 18px; border: 1px solid #bfdbfe; border-left: 6px solid #2563eb; border-radius: 6px; background: #eff6ff; font-size: 14px; line-height: 1.5;">

    <strong>Pontos de controle:</strong> Colunas derivadas criadas, categorias coerentes e base analítica atualizada.

        <td style="padding: 7px 9px; color: #1e40af;">As proxies não equivalem a diagnóstico clínico individual; devem ser interpretadas como marcadores operacionais retrospectivos.</td>
      </tr>

````python
def criar_variaveis_derivadas(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()

    data["NICU"] = data["AB_NICU"].astype(float)

    data["PRETERM"] = np.where(data["COMBGEST"] < 37, 1, np.where(data["COMBGEST"].notna(), 0, np.nan))
    data["LOW_BW"] = np.where(data["DBWT"].notna(), (data["DBWT"] < 2500).astype(int), np.nan)
    data["VERY_LOW_BW"] = np.where(data["DBWT"].notna(), (data["DBWT"] < 1500).astype(int), np.nan)
    data["EXTREME_LOW_BW"] = np.where(data["DBWT"].notna(), (data["DBWT"] < 1000).astype(int), np.nan)

    # Alias em português usados nas tabelas complementares por decil.
    data["BAIXO_PESO"] = data["LOW_BW"]
    data["MUITO_BAIXO_PESO"] = data["VERY_LOW_BW"]
    data["EXTREMO_BAIXO_PESO"] = data["EXTREME_LOW_BW"]

    data["LOW_APGAR5"] = np.where(data["APGAR5"].notna(), (data["APGAR5"] < 7).astype(int), np.nan)

    data["GEMELAR"] = np.where(data["DPLURAL"] == 2, 1, np.where(data["DPLURAL"] == 1, 0, np.nan))
    data["MULTIPLA"] = np.where(data["DPLURAL"] >= 2, 1, np.where(data["DPLURAL"] == 1, 0, np.nan))

    def pluralidade_label(x):
        if pd.isna(x): return np.nan
        if x == 1: return "Gestação única"
        if x == 2: return "Gemelar"
        if x == 3: return "Trigemelar"
        if x >= 4: return "Quádrupla ou mais"
        return np.nan
    data["PLURALIDADE_CAT"] = data["DPLURAL"].apply(pluralidade_label)

    # Grupos usados nas tabelas finais do TCC.
    data["GRUPO_GESTACIONAL_DETALHADO"] = pd.cut(
        data["COMBGEST"],
        bins=[0, 27, 31, 33, 36, 38, 42, 99],
        labels=[
            "Pré-termo extremo",
            "Muito pré-termo",
            "Pré-termo moderado",
            "Pré-termo tardio",
            "Termo precoce",
            "Termo",
            "Pós-termo"
        ],
        right=True
    )

    data["GRUPO_GESTACIONAL_SIMPLES"] = pd.cut(
        data["COMBGEST"],
        bins=[0, 31, 36, 42, 99],
        labels=["Muito pré-termo", "Pré-termo", "Termo", "Pós-termo"],
        right=True
    )

    # Proxy conservadora: maior especificidade analítica.
    # Exige nascimento pré-termo e registro de corioamnionite.
    # Não equivale a diagnóstico clínico individual de RPM/PPROM.
    data["PROM_PROXY_CONSERV"] = np.where(
        (data["PRETERM"] == 1) & (data["LD_CHOR"] == 1),
        1,
        np.where((data["PRETERM"].notna()) & (data["LD_CHOR"].notna()), 0, np.nan)
    )

    # Proxy ampla: maior sensibilidade analítica para contexto de vulnerabilidade pré-termo.
    # Combina prematuridade com corioamnionite, antibioticoterapia intraparto ou esteroides.
    # Quando PRETERM está informado e nenhum componente está positivo, o registro é classificado como 0.
    # Essa decisão operacional permite análise binária, mas pode subestimar casos em situações de ausência
    # ou sub-registro dos componentes LD_CHOR, LD_ANTB e LD_STER.
    data["PROM_PROXY_WIDE"] = np.where(
        (data["PRETERM"] == 1) & ((data["LD_CHOR"] == 1) | (data["LD_ANTB"] == 1) | (data["LD_STER"] == 1)),
        1,
        np.where(data["PRETERM"].notna(), 0, np.nan)
    )

    # Indicador auxiliar de documentação: componentes da proxy ampla simultaneamente ausentes.
    # Não entra nos modelos; serve para auditoria de completude e interpretação metodológica.
    data["PROM_PROXY_WIDE_COMPONENTES_AUSENTES"] = np.where(
        data[["LD_CHOR", "LD_ANTB", "LD_STER"]].isna().all(axis=1),
        1, 0
    )

    data["DESFECHO_NEONATAL_GRAVE"] = np.where(
        (data["AB_NICU"] == 1) |
        (data["AB_AVEN6"] == 1) |
        (data["AB_SEIZ"] == 1) |
        (data["AB_SURF"] == 1) |
        (data["AB_ANTI"] == 1) |
        (data["LOW_APGAR5"] == 1) |
        (data["COMBGEST"] < 32) |
        (data["DBWT"] < 1500),
        1, 0
    )

    # Versão do desfecho grave sem componentes de desfecho neonatal direto, útil para regressão de sensibilidade.
    data["DESFECHO_GRAVE_SEM_COMPONENTES"] = np.where(
        (data["LOW_APGAR5"] == 1) | (data["COMBGEST"] < 32) | (data["DBWT"] < 1500),
        1, 0
    )

    return data

data = criar_variaveis_derivadas(df)
data.to_parquet(DIR_DADOS_FINAIS / "base_analitica_tcc_utin_rpm_pprom.parquet", index=False)
data.shape
````

## 4.1 Contrato metodológico das variáveis derivadas e da proxy operacional

Esta seção transforma em checagem computacional as cautelas descritas no relatório final. A proxy operacional compatível com contexto de RPM/PPROM é útil como marcador analítico, mas não deve ser lida como diagnóstico individual.

O objetivo é reduzir ambiguidades na defesa do trabalho e facilitar a auditoria do pipeline.


<div style="margin: 14px 0 22px 0; padding: 14px 18px; border: 1px solid #bfdbfe; border-left: 6px solid #2563eb; border-radius: 6px; background: #eff6ff; font-size: 14px; line-height: 1.5;">

  <h3 style="margin: 0 0 10px 0; color: #1e3a8a;">
    ℹ️ Contrato metodológico das variáveis derivadas
  </h3>

  <p style="margin: 0 0 12px 0; color: #1e40af;">
    <strong>Descrição técnica:</strong> Esta etapa transforma os limites interpretativos das variáveis derivadas e da proxy operacional em checagens computacionais.
    <strong>Pontos de controle:</strong> Variáveis esperadas verificadas, inconsistências sinalizadas e contrato metodológico exibido.
  </p>

  <table style="border-collapse: collapse; width: 100%; font-size: 13px; margin-top: 10px;">
    <thead>
      <tr style="background-color: #dbeafe; border-bottom: 2px solid #2563eb;">
        <th style="padding: 7px 9px; text-align: left; color: #1e3a8a; width: 28%;">Tópico</th>
        <th style="padding: 7px 9px; text-align: left; color: #1e3a8a;">Descrição</th>
      </tr>
    </thead>
    <tbody>
      <tr style="border-bottom: 1px solid #bfdbfe;">
        <td style="padding: 7px 9px; font-weight: 600; color: #1d4ed8;">O que o código faz</td>
        <td style="padding: 7px 9px; color: #1e40af;">Confere presença de colunas derivadas, calcula indicadores de consistência e registra ressalvas sobre a proxy RPM/PPROM.</td>
      </tr>
      <tr style="border-bottom: 1px solid #bfdbfe;">
        <td style="padding: 7px 9px; font-weight: 600; color: #1d4ed8;">Entradas/insumos</td>
        <td style="padding: 7px 9px; color: #1e40af;">Base analítica após criação das variáveis derivadas.</td>
      </tr>
      <tr style="border-bottom: 1px solid #bfdbfe;">
        <td style="padding: 7px 9px; font-weight: 600; color: #1d4ed8;">Saídas/resultado esperado</td>
        <td style="padding: 7px 9px; color: #1e40af;">Tabela de checagem metodológica e mensagens de validação sobre as variáveis esperadas.</td>
      </tr>
      <tr>
        <td style="padding: 7px 9px; font-weight: 600; color: #1d4ed8;">Observação para o TCC</td>
        <td style="padding: 7px 9px; color: #1e40af;">A célula reforça que a proxy é analítica e retrospectiva, evitando leitura diagnóstica indevida.</td>
      </tr>
    </tbody>

````python
# ============================================================
# Checagem metodológica das variáveis derivadas e da proxy operacional
# ============================================================
# Execute após a criação de `data = criar_variaveis_derivadas(df)`.

VARIAVEIS_DERIVADAS_ESPERADAS = [
    "NICU", "PRETERM", "LOW_BW", "LOW_APGAR5",
    "PROM_PROXY_WIDE", "PROM_PROXY_CONSERV", "GEMELAR"
]

def checar_variaveis_derivadas(base):
    linhas = []
    for var in VARIAVEIS_DERIVADAS_ESPERADAS:
        existe = var in base.columns
        linhas.append({
            "Variável": var,
            "Status": "OK" if existe else "PENDENTE",
            "Não nulos": int(base[var].notna().sum()) if existe else 0,
            "Valores únicos": ", ".join(map(str, sorted(pd.Series(base[var].dropna().unique()).astype(str).head(10)))) if existe else ""
        })
    return pd.DataFrame(linhas)

def checar_coerencia_proxy(base):
    linhas = []
    if {"PROM_PROXY_CONSERV", "PROM_PROXY_WIDE"}.issubset(base.columns):
        conserv_maior_que_wide = int(((base["PROM_PROXY_CONSERV"] == 1) & (base["PROM_PROXY_WIDE"] != 1)).sum())
        linhas.append({
            "Checagem": "Proxy conservadora deve estar contida na proxy ampla",
            "Status": "OK" if conserv_maior_que_wide == 0 else "REVISAR",
            "Detalhe": f"Registros conservadores fora da ampla: {conserv_maior_que_wide}"
        })
    if {"PRETERM", "PROM_PROXY_WIDE"}.issubset(base.columns):
        wide_sem_preterm = int(((base["PROM_PROXY_WIDE"] == 1) & (base["PRETERM"] != 1)).sum())
        linhas.append({
            "Checagem": "Proxy ampla deve ocorrer apenas em registros pré-termo pela regra definida",
            "Status": "OK" if wide_sem_preterm == 0 else "REVISAR",
            "Detalhe": f"Registros proxy ampla sem PRETERM=1: {wide_sem_preterm}"
        })
    if {"PRETERM", "PROM_PROXY_CONSERV"}.issubset(base.columns):
        conserv_sem_preterm = int(((base["PROM_PROXY_CONSERV"] == 1) & (base["PRETERM"] != 1)).sum())
        linhas.append({
            "Checagem": "Proxy conservadora deve ocorrer apenas em registros pré-termo pela regra definida",
            "Status": "OK" if conserv_sem_preterm == 0 else "REVISAR",
            "Detalhe": f"Registros proxy conservadora sem PRETERM=1: {conserv_sem_preterm}"
        })
    return pd.DataFrame(linhas)

if "data" in globals():
    checklist_variaveis_derivadas = checar_variaveis_derivadas(data)
    checklist_coerencia_proxy = checar_coerencia_proxy(data)

    try:
        checklist_variaveis_derivadas.to_csv(
            DIR_MANIFESTO / "checklist_variaveis_derivadas.csv",
            index=False,
            encoding="utf-8-sig"
        )
        checklist_coerencia_proxy.to_csv(
            DIR_MANIFESTO / "checklist_coerencia_proxy_operacional.csv",
            index=False,
            encoding="utf-8-sig"
        )
    except Exception as exc:
        print("Aviso: não foi possível exportar checklists neste momento:", exc)

    display(Markdown("### Checklist das variáveis derivadas"))
    display(checklist_variaveis_derivadas)

    display(Markdown("### Coerência interna da proxy operacional"))
    display(checklist_coerencia_proxy)
else:
    display(Markdown(
        "Execute primeiro a célula que cria `data = criar_variaveis_derivadas(df)` para habilitar esta checagem."
    ))
````

  <strong style="color: #166534; font-size: 12px; text-transform: uppercase; letter-spacing: 0.5px;">✅ Resultado esperado</strong>

````python
from matplotlib.ticker import PercentFormatter
from matplotlib.colors import ListedColormap, BoundaryNorm
import matplotlib.patches as mpatches

# ============================================================
# Padrão visual unificado dos gráficos do TCC
# Adaptado do notebook 08_graficos_finais_eda_tcc_unificado_estilo_suave_CORRIGIDO.ipynb
# ============================================================

STYLE = {
    "primary": "#AFCBEA",
    "secondary": "#BFDDB8",
    "accent": "#F6C99B",
    "risk": "#E8A6A6",
    "neutral": "#D6D6D6",
    "purple": "#C7B8EA",
    "yellow": "#F3E7A3",
    "text": "#2F3A45",
    "grid": "#E6EAF0",
    "spine": "#CBD5E1",
    "white": "#FFFFFF",
}

BAR_COLORS = [
    STYLE["primary"],
    STYLE["secondary"],
    STYLE["accent"],
    STYLE["purple"],
    STYLE["risk"],
    STYLE["yellow"],
    STYLE["neutral"],
]

BINARY_COLORS = {
    "Sem UTIN": STYLE["secondary"],
    "Com UTIN": STYLE["risk"],
    "Não pré-termo": STYLE["secondary"],
    "Pré-termo": STYLE["risk"],
    "Não prematuro": STYLE["secondary"],
    "Prematuro": STYLE["risk"],
    "Gestação única": STYLE["primary"],
    "Gestação gemelar": STYLE["accent"],
    "Única": STYLE["primary"],
    "Gemelar": STYLE["accent"],
    "Trigêmeos": STYLE["purple"],
    "Trigemelar": STYLE["purple"],
    "Quádruplos": STYLE["risk"],
    "Quádrupla ou mais": STYLE["risk"],
    "Quíntuplos ou mais": STYLE["yellow"],
    "Ignorado": STYLE["neutral"],
    "Sem proxy RPM/PPROM": STYLE["secondary"],
    "Com proxy RPM/PPROM": STYLE["risk"],
    "Sem proxy conservadora": STYLE["secondary"],
    "Com proxy conservadora": STYLE["risk"],
    "Sem proxy ampla": STYLE["secondary"],
    "Com proxy ampla": STYLE["risk"],
    "Ausente": STYLE["secondary"],
    "Presente": STYLE["risk"],
}

plt.rcParams.update({
    "figure.facecolor": STYLE["white"],
    "axes.facecolor": STYLE["white"],
    "axes.edgecolor": STYLE["spine"],
    "axes.labelcolor": STYLE["text"],
    "xtick.color": STYLE["text"],
    "ytick.color": STYLE["text"],
    "text.color": STYLE["text"],
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "font.size": 10,
})

def tabela_distribuicao(df, var, target="NICU"):
    tab = (
        df.dropna(subset=[var, target])
        .groupby(var, observed=True)[target]
        .agg(total_registros="count", casos_utin="sum", taxa_utin="mean")
        .reset_index()
    )
    tab["taxa_utin_pct"] = tab["taxa_utin"] * 100
    return tab

def tabela_cruzada_percentual(df, linhas, colunas, target="NICU"):
    tab = (
        df.dropna(subset=linhas + colunas + [target])
        .groupby(linhas + colunas, observed=True)[target]
        .agg(total="count", com_utin="sum")
        .reset_index()
    )
    tab["sem_utin"] = tab["total"] - tab["com_utin"]
    tab["pct_sem_utin"] = tab["sem_utin"] / tab["total"] * 100
    tab["pct_com_utin"] = tab["com_utin"] / tab["total"] * 100
    return tab

def format_pct(value, decimals=1):
    if pd.isna(value):
        return ""
    return f"{value:.{decimals}f}%".replace(".", ",")

def _bar_colors(labels, default_palette=None):
    if default_palette is None:
        default_palette = BAR_COLORS
    colors = []
    for i, label in enumerate(labels):
        label_str = str(label)
        colors.append(BINARY_COLORS.get(label_str, default_palette[i % len(default_palette)]))
    return colors

def apply_academic_style(ax, title, xlabel, ylabel, rotate_x=25, grid_axis="y"):
    ax.set_title(title, pad=14)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)

    if grid_axis:
        ax.grid(axis=grid_axis, linestyle="-", linewidth=0.8, alpha=0.45, color=STYLE["grid"])
        ax.set_axisbelow(True)

    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)

    ax.spines["left"].set_color(STYLE["spine"])
    ax.spines["bottom"].set_color(STYLE["spine"])

    if rotate_x:
        plt.setp(ax.get_xticklabels(), rotation=rotate_x, ha="right")
    else:
        plt.setp(ax.get_xticklabels(), rotation=0, ha="center")

def add_bar_labels(ax, bars, values, suffix="%", decimals=1, offset=None):
    if offset is None:
        max_val = max([float(v) for v in values if not pd.isna(v)] + [1])
        offset = max_val * 0.02

    for bar, value in zip(bars, values):
        if pd.isna(value):
            continue
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + offset,
            f"{value:.{decimals}f}{suffix}".replace(".", ","),
            ha="center",
            va="bottom",
            fontsize=9,
            color=STYLE["text"],
        )

def salvar_figura(fig, arquivo):
    caminho = DIR_GRAFICOS / arquivo
    fig.tight_layout()
    fig.savefig(caminho, dpi=300, bbox_inches="tight")
    plt.show()
    return caminho

def grafico_barras_taxa(
    tab,
    x,
    y="taxa_utin_pct",
    titulo="",
    arquivo="grafico.png",
    rotacao=25,
    xlabel="",
    ylabel="Taxa de internação em UTIN (%)",
    min_ylim=10,
):
    labels = tab[x].astype(str)
    values = tab[y].astype(float)

    fig, ax = plt.subplots(figsize=(9, 5.5), facecolor=STYLE["white"])
    bars = ax.bar(
        labels,
        values,
        color=_bar_colors(labels),
        edgecolor=STYLE["white"],
        linewidth=1.2,
        width=0.68,
    )

    apply_academic_style(ax, title=titulo, xlabel=xlabel, ylabel=ylabel, rotate_x=rotacao)
    ax.yaxis.set_major_formatter(PercentFormatter(xmax=100))
    ymax = max(float(values.max()) * 1.20, min_ylim) if len(values) else min_ylim
    ax.set_ylim(0, ymax)
    add_bar_labels(ax, bars, values, suffix="%", decimals=1)

    return salvar_figura(fig, arquivo)

def grafico_barras_agrupadas(
    tab,
    group_col,
    value_cols,
    labels,
    titulo,
    arquivo,
    xlabel="",
    ylabel="Proporção (%)",
    rotacao=25,
):
    x = np.arange(len(tab))
    width = 0.36 if len(value_cols) == 2 else 0.26
    offsets = np.linspace(
        -width * (len(value_cols) - 1) / 2,
        width * (len(value_cols) - 1) / 2,
        len(value_cols),
    )

    fig, ax = plt.subplots(figsize=(10, 5.8), facecolor=STYLE["white"])
    all_values = []

    for idx, (col, label) in enumerate(zip(value_cols, labels)):
        values = tab[col].astype(float).values
        all_values.extend(values.tolist())

        bars = ax.bar(
            x + offsets[idx],
            values,
            width,
            label=label,
            color=BAR_COLORS[idx % len(BAR_COLORS)],
            edgecolor=STYLE["white"],
            linewidth=1.2,
        )
        add_bar_labels(ax, bars, values, suffix="%", decimals=1)

    ax.set_xticks(x)
    ax.set_xticklabels(tab[group_col].astype(str))
    apply_academic_style(ax, title=titulo, xlabel=xlabel, ylabel=ylabel, rotate_x=rotacao)
    ax.yaxis.set_major_formatter(PercentFormatter(xmax=100))
    ax.set_ylim(0, max(max(all_values) * 1.25, 10) if all_values else 10)
    ax.legend(frameon=False)

    return salvar_figura(fig, arquivo)

def grafico_linhas_taxa(
    tab,
    x,
    y,
    grupo,
    titulo,
    arquivo,
    xlabel="",
    ylabel="Taxa de internação em UTIN (%)",
    rotacao=25,
):
    fig, ax = plt.subplots(figsize=(10, 5.5), facecolor=STYLE["white"])
    for idx, (cat, sub) in enumerate(tab.groupby(grupo, observed=True)):
        ax.plot(
            sub[x],
            sub[y],
            marker="o",
            linewidth=1.8,
            label=str(cat),
            color=BAR_COLORS[idx % len(BAR_COLORS)],
        )

    apply_academic_style(ax, title=titulo, xlabel=xlabel, ylabel=ylabel, rotate_x=rotacao)
    ax.yaxis.set_major_formatter(PercentFormatter(xmax=100))
    ax.legend(frameon=False)
    return salvar_figura(fig, arquivo)

def grafico_boxplot_peso_utin(data, titulo, arquivo):
    tmp = data.dropna(subset=["NICU", "DBWT"]).copy()
    tmp["NICU_LABEL"] = tmp["NICU"].map({0: "Sem UTIN", 1: "Com UTIN"})

    fig, ax = plt.subplots(figsize=(8.5, 5.5), facecolor=STYLE["white"])
    tmp.boxplot(
        column="DBWT",
        by="NICU_LABEL",
        ax=ax,
        patch_artist=True,
        boxprops=dict(facecolor=STYLE["primary"], color=STYLE["spine"]),
        medianprops=dict(color=STYLE["risk"], linewidth=1.5),
        whiskerprops=dict(color=STYLE["spine"]),
        capprops=dict(color=STYLE["spine"]),
        flierprops=dict(marker=".", markersize=2, markerfacecolor=STYLE["neutral"], markeredgecolor=STYLE["neutral"], alpha=0.25),
    )
    plt.suptitle("")
    apply_academic_style(
        ax,
        title=titulo,
        xlabel="Internação em UTIN",
        ylabel="Peso ao nascer (g)",
        rotate_x=0,
    )
    return salvar_figura(fig, arquivo)

def grafico_heatmap_manchester(heat, titulo, arquivo):
    # Escala inspirada no Protocolo de Manchester usada no notebook 08:
    # 0–10% muito baixo; 10–25% baixo; 25–50% intermediário; 50–75% alto; 75–100% muito alto.
    manchester_colors = [
        "#4A90E2",  # azul
        "#7BC96F",  # verde
        "#F7D154",  # amarelo
        "#F5A142",  # laranja
        "#D9534F",  # vermelho
    ]
    bounds = [0, 10, 25, 50, 75, 100]
    cmap_manchester = ListedColormap(manchester_colors)
    norm_manchester = BoundaryNorm(bounds, cmap_manchester.N)

    fig, ax = plt.subplots(figsize=(10.5, 5.8), facecolor=STYLE["white"])
    imagem = ax.imshow(heat.values, aspect="auto", cmap=cmap_manchester, norm=norm_manchester)

    ax.set_title(titulo, pad=14, fontweight="bold")
    ax.set_xlabel("Tipo de gestação")
    ax.set_ylabel("Grupo gestacional")
    ax.set_xticks(np.arange(len(heat.columns)))
    ax.set_xticklabels(heat.columns)
    ax.set_yticks(np.arange(len(heat.index)))
    ax.set_yticklabels(heat.index)

    for i in range(len(heat.index)):
        for j in range(len(heat.columns)):
            valor = heat.values[i, j]
            if not np.isnan(valor):
                text_color = "white" if valor >= 50 else STYLE["text"]
                ax.text(
                    j, i, format_pct(valor),
                    ha="center", va="center",
                    fontsize=9, fontweight="bold", color=text_color,
                )

    for spine in ["top", "right", "left", "bottom"]:
        ax.spines[spine].set_visible(False)

    ax.set_xticks(np.arange(-0.5, len(heat.columns), 1), minor=True)
    ax.set_yticks(np.arange(-0.5, len(heat.index), 1), minor=True)
    ax.grid(which="minor", color=STYLE["white"], linestyle="-", linewidth=2)
    ax.tick_params(which="minor", bottom=False, left=False)

    cbar = plt.colorbar(
        imagem,
        ax=ax,
        boundaries=bounds,
        ticks=bounds,
        spacing="proportional",
        fraction=0.045,
        pad=0.04,
    )
    cbar.set_label("Taxa de internação em UTIN (%)")

    fig.subplots_adjust(right=0.62)

    legenda_manchester = [
        mpatches.Patch(color="#4A90E2", label="0–10%: muito baixo"),
        mpatches.Patch(color="#7BC96F", label="10–25%: baixo"),
        mpatches.Patch(color="#F7D154", label="25–50%: intermediário"),
        mpatches.Patch(color="#F5A142", label="50–75%: alto"),
        mpatches.Patch(color="#D9534F", label="75–100%: muito alto"),
    ]
    ax.legend(
        handles=legenda_manchester,
        title="Faixas de risco",
        loc="upper left",
        bbox_to_anchor=(1.25, 1.0),
        frameon=False,
        borderaxespad=0.0,
    )

    caminho = DIR_GRAFICOS / arquivo
    fig.savefig(caminho, dpi=300, bbox_inches="tight")
    plt.show()
    return caminho

def salvar_tabela(numero, titulo, df):
    return salvar_df(df, DIR_TABELAS, numero, titulo, "tabela")

def salvar_quadro(numero, titulo, df):
    return salvar_df(df, DIR_QUADROS, numero, titulo, "quadro")
````

  <strong style="color: #166534; font-size: 12px; text-transform: uppercase; letter-spacing: 0.5px;">✅ Resultado esperado</strong>

      <td style="padding: 5px 8px; color: #14532d;">Funções utilitárias e padrão visual disponíveis.</td>

````python
# Tabela 1
tabela_1 = pd.DataFrame([
    ["Base", "Natality Data 2024 Public Use File"],
    ["Tipo", "Microdados públicos de nascimentos"],
    ["Formato original", "Fixed-width"],
    ["Unidade de análise", "Nascimento/recém-nascido registrado"],
    ["Desfecho principal", "AB_NICU/NICU: internação em UTIN"],
    ["Finalidade", "Estratificação retrospectiva e análise observacional"],
], columns=["Item", "Descrição"])
salvar_tabela(1, TABELAS[0][1], tabela_1)

# Tabela 2
tabela_2 = pd.DataFrame([
    ["PROM/PPROM diagnóstico direto", "Não disponível no arquivo público", "Exigiu aproximação operacional"],
    ["LD_CHOR", "Disponível", "Componente da proxy conservadora/ampla"],
    ["LD_ANTB", "Disponível", "Componente da proxy ampla"],
    ["LD_STER", "Disponível", "Componente da proxy ampla"],
    ["PRETERM/COMBGEST", "Disponível/derivável", "Restrição gestacional da proxy"],
], columns=["Informação", "Disponibilidade", "Uso no TCC"])
salvar_tabela(2, TABELAS[1][1], tabela_2)

# Tabela 3
# Classificação explícita do papel analítico das variáveis originais.
# Evita a categoria ambígua "preditor/desfecho" e deixa claro que AB_NICU é o desfecho principal.
papeis_variaveis_originais = {
    "AB_NICU": "desfecho principal",
    "AB_AVEN1": "variável neonatal/perinatal; excluída do cenário restritivo por proximidade temporal",
    "AB_AVEN6": "variável neonatal/perinatal; excluída do cenário restritivo por proximidade temporal",
    "AB_SURF": "variável neonatal/perinatal; excluída do cenário restritivo por proximidade temporal",
    "AB_ANTI": "variável neonatal/perinatal; excluída do cenário restritivo por proximidade temporal",
    "AB_SEIZ": "variável neonatal/perinatal; excluída do cenário restritivo por proximidade temporal",
    "APGAR5": "variável neonatal/perinatal; excluída do cenário restritivo por proximidade temporal",
    "COMBGEST": "preditor perinatal/gestacional e base para variáveis derivadas",
    "DBWT": "preditor perinatal/neonatal e base para variáveis derivadas",
    "DPLURAL": "preditor obstétrico; base para gemelaridade/pluralidade",
    "SETORDER_R": "preditor obstétrico; ordem do nascimento em gestação múltipla",
    "LD_CHOR": "preditor obstétrico; componente de proxy operacional",
    "LD_ANTB": "preditor obstétrico/assistencial; componente de proxy operacional",
    "LD_STER": "preditor obstétrico/assistencial; componente de proxy operacional",
    "LD_INDL": "preditor obstétrico/assistencial",
    "MAGER": "preditor materno",
    "PREVIS": "preditor de cuidado pré-natal",
    "WIC": "preditor socioassistencial",
    "RF_GDIAB": "preditor materno/obstétrico",
    "RF_GHYPE": "preditor materno/obstétrico",
    "GESTREC3": "variável gestacional recodificada",
    "BWTR4": "variável de peso ao nascer recodificada",
}

tabela_3 = pd.DataFrame([
    [c, "variável original selecionada do arquivo Natality 2024", papeis_variaveis_originais.get(c, "preditor/variável de apoio")]
    for c in COLSPECS.keys()
], columns=["Variável", "Descrição operacional", "Papel analítico"])
salvar_tabela(3, TABELAS[2][1], tabela_3)

# Tabela 4
tabela_4 = pd.DataFrame([
    ["NICU", "AB_NICU recodificada", "Desfecho principal"],
    ["PRETERM", "COMBGEST < 37", "Prematuridade"],
    ["LOW_BW", "DBWT < 2500g", "Baixo peso"],
    ["VERY_LOW_BW", "DBWT < 1500g", "Muito baixo peso"],
    ["LOW_APGAR5", "APGAR5 < 7", "Vitalidade neonatal"],
    ["GEMELAR", "DPLURAL == 2", "Gemelaridade"],
    ["MULTIPLA", "DPLURAL >= 2", "Gestação múltipla"],
    ["PROM_PROXY_CONSERV", "PRETERM = 1 e LD_CHOR = 1", "Proxy conservadora; maior especificidade analítica"],
    ["PROM_PROXY_WIDE", "PRETERM = 1 e (LD_CHOR = 1 ou LD_ANTB = 1 ou LD_STER = 1)", "Proxy ampla; marcador operacional ampliado de vulnerabilidade pré-termo"],
    ["PROM_PROXY_WIDE_COMPONENTES_AUSENTES", "LD_CHOR, LD_ANTB e LD_STER ausentes simultaneamente", "Auditoria de completude; não utilizado como preditor"],
    ["DESFECHO_NEONATAL_GRAVE", "Composto neonatal/gestacional", "Desfecho secundário"],
], columns=["Variável derivada", "Critério", "Uso analítico"])
salvar_tabela(4, TABELAS[3][1], tabela_4)

# Tabela 5
# Síntese alinhada ao documento de referência renumerado:
# separa base processada/carregada e amostra analítica final nas variáveis-chave.
VARIAVEIS_CHAVE_AMOSTRA_ANALITICA = [
    "NICU", "COMBGEST", "DBWT", "DPLURAL", "GEMELAR",
    "PRETERM", "LOW_BW", "PROM_PROXY_WIDE", "PROM_PROXY_CONSERV"
]
VARIAVEIS_CHAVE_AMOSTRA_ANALITICA = [c for c in VARIAVEIS_CHAVE_AMOSTRA_ANALITICA if c in data.columns]

n_base_processada = int(len(data))
n_amostra_analitica_final = int(data.dropna(subset=VARIAVEIS_CHAVE_AMOSTRA_ANALITICA).shape[0])
n_excluidos_incompletude = int(n_base_processada - n_amostra_analitica_final)
pct_excluidos_incompletude = (
    n_excluidos_incompletude / n_base_processada * 100 if n_base_processada else np.nan
)

tabela_5 = pd.DataFrame([
    ["Base processada/carregada", n_base_processada],
    ["Amostra analítica final nas variáveis-chave", n_amostra_analitica_final],
    ["Registros excluídos por incompletude nas variáveis-chave", n_excluidos_incompletude],
    ["Percentual excluído por incompletude nas variáveis-chave (%)", pct_excluidos_incompletude],
    ["Taxa de UTIN", data["NICU"].mean()],
    ["Prematuridade", data["PRETERM"].mean()],
    ["Baixo peso", data["LOW_BW"].mean()],
    ["Proxy ampla", data["PROM_PROXY_WIDE"].mean()],
    ["Proxy conservadora", data["PROM_PROXY_CONSERV"].mean()],
    ["Gemelaridade", data["GEMELAR"].mean()],
], columns=["Indicador", "Valor"])
salvar_tabela(5, TABELAS[4][1], tabela_5)

# Tabela 6
tabela_6 = pd.DataFrame({
    "Indicador": ["Registros", "Colunas", "AB_NICU válido", "NICU positivos", "NICU negativos"],
    "Valor": [len(data), data.shape[1], data["AB_NICU"].notna().sum(), int(data["NICU"].sum()), int((data["NICU"]==0).sum())]
})
salvar_tabela(6, TABELAS[5][1], tabela_6)

# Tabela 7
tabela_7 = data[list(COLSPECS.keys())].isna().mean().mul(100).sort_values(ascending=False).reset_index()
tabela_7.columns = ["Variável", "Percentual ausente"]
salvar_tabela(7, TABELAS[6][1], tabela_7)

# Tabela 8
tabela_8 = data["AB_NICU"].value_counts(dropna=False).rename_axis("AB_NICU").reset_index(name="n")
tabela_8["percentual"] = tabela_8["n"] / tabela_8["n"].sum() * 100
salvar_tabela(8, TABELAS[7][1], tabela_8)

# Tabelas 9 a 21
salvar_tabela(9, TABELAS[8][1], tabela_distribuicao(data, "GEMELAR"))
salvar_tabela(10, TABELAS[9][1], tabela_distribuicao(data, "GEMELAR", target="DESFECHO_NEONATAL_GRAVE"))
salvar_tabela(11, TABELAS[10][1], tabela_cruzada_percentual(data, ["PLURALIDADE_CAT"], ["PROM_PROXY_WIDE"]))
salvar_tabela(12, TABELAS[11][1], tabela_distribuicao(data, "PRETERM"))
salvar_tabela(13, TABELAS[12][1], tabela_distribuicao(data, "GRUPO_GESTACIONAL_DETALHADO"))

# Tabela 14: correlação
corr = data[["COMBGEST", "DBWT"]].dropna().corr().loc["COMBGEST", "DBWT"]
tabela_14 = pd.DataFrame([["COMBGEST e DBWT", corr]], columns=["Variáveis", "Correlação"])
salvar_tabela(14, TABELAS[13][1], tabela_14)

# Tabela 15: VIF
if VIF_AVAILABLE:
    vif_df = data[["COMBGEST", "DBWT"]].dropna()
    X_vif = pd.DataFrame(SimpleImputer(strategy="median").fit_transform(vif_df), columns=vif_df.columns)
    X_vif["const"] = 1
    tabela_15 = pd.DataFrame({
        "Variável": ["COMBGEST", "DBWT"],
        "VIF": [
            variance_inflation_factor(X_vif[["const", "COMBGEST", "DBWT"]].values, 1),
            variance_inflation_factor(X_vif[["const", "COMBGEST", "DBWT"]].values, 2),
        ]
    })
else:
    tabela_15 = pd.DataFrame({"Variável": ["COMBGEST", "DBWT"], "VIF": [np.nan, np.nan]})
salvar_tabela(15, TABELAS[14][1], tabela_15)

salvar_tabela(16, TABELAS[15][1], tabela_cruzada_percentual(data, ["GEMELAR"], ["PROM_PROXY_WIDE"]))

tabela_17 = pd.DataFrame([
    ["PROM_PROXY_CONSERV", data["PROM_PROXY_CONSERV"].mean(), data.loc[data["PROM_PROXY_CONSERV"]==1, "NICU"].mean()],
    ["PROM_PROXY_WIDE", data["PROM_PROXY_WIDE"].mean(), data.loc[data["PROM_PROXY_WIDE"]==1, "NICU"].mean()],
], columns=["Proxy", "Prevalência", "Taxa de UTIN entre positivos"])
salvar_tabela(17, TABELAS[16][1], tabela_17)

# Tabela 18: distribuição percentual dos casos de UTIN segundo pluralidade e proxy ampla.
tabela_18 = (
    data.dropna(subset=["PLURALIDADE_CAT", "PROM_PROXY_WIDE", "NICU"])
    .groupby(["PLURALIDADE_CAT", "PROM_PROXY_WIDE"], observed=True)["NICU"]
    .agg(total_registros="count", casos_utin="sum", taxa_utin="mean")
    .reset_index()
)
total_casos_utin_t18 = tabela_18["casos_utin"].sum()
tabela_18["taxa_utin_pct"] = tabela_18["taxa_utin"] * 100
tabela_18["distribuicao_percentual_dos_casos_utin"] = np.where(
    total_casos_utin_t18 > 0,
    tabela_18["casos_utin"] / total_casos_utin_t18 * 100,
    np.nan
)
salvar_tabela(18, TABELAS[17][1], tabela_18)
salvar_tabela(19, TABELAS[18][1], tabela_cruzada_percentual(data, ["GRUPO_GESTACIONAL_SIMPLES"], ["PROM_PROXY_CONSERV"]))
salvar_tabela(20, TABELAS[19][1], tabela_cruzada_percentual(data, ["GRUPO_GESTACIONAL_SIMPLES"], ["PROM_PROXY_WIDE"]))

# Tabela 21: grupos com maiores taxas observadas
tabela_21 = (
    data.dropna(subset=["GRUPO_GESTACIONAL_DETALHADO", "PLURALIDADE_CAT", "PROM_PROXY_WIDE", "NICU"])
    .groupby(["GRUPO_GESTACIONAL_DETALHADO", "PLURALIDADE_CAT", "PROM_PROXY_WIDE"], observed=True)["NICU"]
    .agg(total_registros="count", casos_utin="sum", taxa_utin="mean")
    .reset_index()
)
tabela_21 = tabela_21[tabela_21["total_registros"] >= 30].sort_values("taxa_utin", ascending=False).head(20)
tabela_21["taxa_utin_pct"] = tabela_21["taxa_utin"] * 100
salvar_tabela(21, TABELAS[20][1], tabela_21)

print("Tabelas 1 a 21 geradas.")
````

````python
# ============================================================
# Gráficos 1 a 13: análise exploratória
# Padrão visual harmonizado com o notebook 08.
# ============================================================

plot_data = data.copy()

# Rótulos de exibição usados apenas nos gráficos.
plot_data["NICU_LABEL"] = plot_data["NICU"].map({0: "Sem UTIN", 1: "Com UTIN"})
plot_data["PRETERM_LABEL"] = plot_data["PRETERM"].map({0: "Não pré-termo", 1: "Pré-termo"})
plot_data["GEMELAR_LABEL"] = plot_data["GEMELAR"].map({0: "Gestação única", 1: "Gestação gemelar"})
plot_data["PROM_PROXY_WIDE_LABEL"] = plot_data["PROM_PROXY_WIDE"].map({
    0: "Sem proxy RPM/PPROM",
    1: "Com proxy RPM/PPROM",
})
plot_data["PROM_PROXY_CONSERV_LABEL"] = plot_data["PROM_PROXY_CONSERV"].map({
    0: "Sem proxy conservadora",
    1: "Com proxy conservadora",
})

def pluralidade_label_grafico(x):
    if pd.isna(x):
        return np.nan
    if x == 1:
        return "Única"
    if x == 2:
        return "Gemelar"
    if x == 3:
        return "Trigêmeos"
    if x == 4:
        return "Quádruplos"
    if x >= 5:
        return "Quíntuplos ou mais"
    return "Ignorado"

plot_data["PLURALIDADE_LABEL"] = plot_data["DPLURAL"].apply(pluralidade_label_grafico)
plot_data["GEST_GROUP"] = plot_data["GRUPO_GESTACIONAL_DETALHADO"]

ordem_gest = [
    "Pré-termo extremo",
    "Muito pré-termo",
    "Pré-termo moderado",
    "Pré-termo tardio",
    "Termo precoce",
    "Termo",
    "Pós-termo",
]

# ------------------------------------------------------------
# Gráfico 1 — Distribuição da variável de desfecho
# ------------------------------------------------------------
tab_g1 = (
    plot_data["NICU_LABEL"]
    .dropna()
    .value_counts()
    .rename_axis("classe")
    .reset_index(name="quantidade")
)
tab_g1["ordem"] = tab_g1["classe"].map({"Sem UTIN": 0, "Com UTIN": 1})
tab_g1 = tab_g1.sort_values("ordem").drop(columns="ordem")
tab_g1["percentual"] = tab_g1["quantidade"] / tab_g1["quantidade"].sum() * 100

fig, ax = plt.subplots(figsize=(8, 5.5), facecolor=STYLE["white"])
labels = tab_g1["classe"].astype(str)
values = tab_g1["percentual"].astype(float)
bars = ax.bar(
    labels,
    values,
    color=_bar_colors(labels),
    edgecolor=STYLE["white"],
    linewidth=1.2,
    width=0.68,
)
apply_academic_style(
    ax,
    title=GRAFICOS[0][1],
    xlabel="Classe do desfecho",
    ylabel="Proporção (%)",
    rotate_x=0,
)
ax.yaxis.set_major_formatter(PercentFormatter(xmax=100))
ax.set_ylim(0, max(float(values.max()) * 1.15, 10))
add_bar_labels(ax, bars, values, suffix="%", decimals=1)
salvar_figura(fig, "grafico_01_distribuicao_ab_nicu.png")

# ------------------------------------------------------------
# Gráfico 2 — Taxa de UTIN segundo gemelaridade
# ------------------------------------------------------------
tab_g2 = tabela_distribuicao(plot_data, "GEMELAR_LABEL")
tab_g2["ordem"] = tab_g2["GEMELAR_LABEL"].map({"Gestação única": 0, "Gestação gemelar": 1})
tab_g2 = tab_g2.sort_values("ordem").drop(columns="ordem")
grafico_barras_taxa(
    tab_g2,
    "GEMELAR_LABEL",
    titulo=GRAFICOS[1][1],
    arquivo="grafico_02_utin_gemelaridade.png",
    rotacao=0,
    xlabel="Tipo de gestação",
)

# ------------------------------------------------------------
# Gráfico 3 — Taxa de UTIN segundo pluralidade gestacional
# ------------------------------------------------------------
tab_g3 = tabela_distribuicao(plot_data, "PLURALIDADE_LABEL")
ordem_plural = ["Única", "Gemelar", "Trigêmeos", "Quádruplos", "Quíntuplos ou mais", "Ignorado"]
tab_g3["PLURALIDADE_LABEL"] = pd.Categorical(
    tab_g3["PLURALIDADE_LABEL"],
    categories=ordem_plural,
    ordered=True,
)
tab_g3 = tab_g3.sort_values("PLURALIDADE_LABEL")
grafico_barras_taxa(
    tab_g3,
    "PLURALIDADE_LABEL",
    titulo=GRAFICOS[2][1],
    arquivo="grafico_03_utin_pluralidade.png",
    rotacao=20,
    xlabel="Pluralidade",
)

# ------------------------------------------------------------
# Gráfico 4 — Taxa de UTIN segundo prematuridade
# ------------------------------------------------------------
tab_g4 = tabela_distribuicao(plot_data, "PRETERM_LABEL")
tab_g4["ordem"] = tab_g4["PRETERM_LABEL"].map({"Não pré-termo": 0, "Pré-termo": 1})
tab_g4 = tab_g4.sort_values("ordem").drop(columns="ordem")
grafico_barras_taxa(
    tab_g4,
    "PRETERM_LABEL",
    titulo=GRAFICOS[3][1],
    arquivo="grafico_04_utin_prematuridade.png",
    rotacao=0,
    xlabel="Prematuridade",
)

# ------------------------------------------------------------
# Gráfico 5 — Taxa de UTIN segundo proxy operacional ampla
# ------------------------------------------------------------
tab_g5 = tabela_distribuicao(plot_data, "PROM_PROXY_WIDE_LABEL")
tab_g5["ordem"] = tab_g5["PROM_PROXY_WIDE_LABEL"].map({
    "Sem proxy RPM/PPROM": 0,
    "Com proxy RPM/PPROM": 1,
})
tab_g5 = tab_g5.sort_values("ordem").drop(columns="ordem")
grafico_barras_taxa(
    tab_g5,
    "PROM_PROXY_WIDE_LABEL",
    titulo=GRAFICOS[4][1],
    arquivo="grafico_05_utin_proxy_ampla.png",
    rotacao=0,
    xlabel="Classificação pela proxy operacional",
)

# ------------------------------------------------------------
# Gráfico 6 — Proporção de prematuridade segundo proxy operacional ampla
# ------------------------------------------------------------
tab_g6 = tabela_distribuicao(
    plot_data.rename(columns={"PRETERM": "OUTCOME_TMP"}),
    "PROM_PROXY_WIDE_LABEL",
    target="OUTCOME_TMP",
).rename(columns={"casos_utin": "casos_prematuridade", "taxa_utin": "taxa_prematuridade"})
tab_g6["ordem"] = tab_g6["PROM_PROXY_WIDE_LABEL"].map({
    "Sem proxy RPM/PPROM": 0,
    "Com proxy RPM/PPROM": 1,
})
tab_g6 = tab_g6.sort_values("ordem").drop(columns="ordem")
grafico_barras_taxa(
    tab_g6,
    "PROM_PROXY_WIDE_LABEL",
    titulo=GRAFICOS[5][1],
    arquivo="grafico_06_prematuridade_proxy.png",
    rotacao=0,
    xlabel="Classificação pela proxy operacional",
    ylabel="Prematuridade (%)",
)

# ------------------------------------------------------------
# Gráfico 7 — Proporção de baixo peso segundo proxy operacional ampla
# ------------------------------------------------------------
tab_g7 = tabela_distribuicao(
    plot_data.rename(columns={"LOW_BW": "OUTCOME_TMP"}),
    "PROM_PROXY_WIDE_LABEL",
    target="OUTCOME_TMP",
).rename(columns={"casos_utin": "casos_baixo_peso", "taxa_utin": "taxa_baixo_peso"})
tab_g7["ordem"] = tab_g7["PROM_PROXY_WIDE_LABEL"].map({
    "Sem proxy RPM/PPROM": 0,
    "Com proxy RPM/PPROM": 1,
})
tab_g7 = tab_g7.sort_values("ordem").drop(columns="ordem")
grafico_barras_taxa(
    tab_g7,
    "PROM_PROXY_WIDE_LABEL",
    titulo=GRAFICOS[6][1],
    arquivo="grafico_07_baixo_peso_proxy.png",
    rotacao=0,
    xlabel="Classificação pela proxy operacional",
    ylabel="Baixo peso ao nascer (%)",
)

# ------------------------------------------------------------
# Gráfico 8 — Prematuridade e baixo peso por grupo de risco e proxy
# Melhoria aplicada do notebook 08: quatro grupos combinando gemelaridade e proxy.
# ------------------------------------------------------------
plot_data["GRUPO_RISCO"] = np.select(
    [
        (plot_data["PROM_PROXY_WIDE"] == 0) & (plot_data["GEMELAR"] == 0),
        (plot_data["PROM_PROXY_WIDE"] == 1) & (plot_data["GEMELAR"] == 0),
        (plot_data["PROM_PROXY_WIDE"] == 0) & (plot_data["GEMELAR"] == 1),
        (plot_data["PROM_PROXY_WIDE"] == 1) & (plot_data["GEMELAR"] == 1),
    ],
    [
        "Única sem proxy RPM/PPROM",
        "Única com proxy RPM/PPROM",
        "Gemelar sem proxy RPM/PPROM",
        "Gemelar com proxy RPM/PPROM",
    ],
    default=None,
)

tab_g8 = (
    plot_data.dropna(subset=["GRUPO_RISCO", "PRETERM", "LOW_BW"])
    .groupby("GRUPO_RISCO", observed=True)[["PRETERM", "LOW_BW"]]
    .mean()
    .mul(100)
    .reset_index()
    .rename(columns={"PRETERM": "prematuridade_pct", "LOW_BW": "baixo_peso_pct"})
)

ordem_risco = [
    "Única sem proxy RPM/PPROM",
    "Única com proxy RPM/PPROM",
    "Gemelar sem proxy RPM/PPROM",
    "Gemelar com proxy RPM/PPROM",
]
tab_g8["GRUPO_RISCO"] = pd.Categorical(tab_g8["GRUPO_RISCO"], categories=ordem_risco, ordered=True)
tab_g8 = tab_g8.sort_values("GRUPO_RISCO")

grafico_barras_agrupadas(
    tab_g8,
    group_col="GRUPO_RISCO",
    value_cols=["prematuridade_pct", "baixo_peso_pct"],
    labels=["Prematuridade", "Baixo peso ao nascer"],
    titulo=GRAFICOS[7][1],
    arquivo="grafico_08_prematuridade_baixo_peso_proxy.png",
    xlabel="Grupo de risco",
    ylabel="Proporção (%)",
    rotacao=25,
)

# ------------------------------------------------------------
# Gráfico 9 — Taxa de UTIN por grupo gestacional e tipo de gestação
# Melhoria aplicada do notebook 08: heatmap com escala em faixas de risco.
# ------------------------------------------------------------
heat = (
    plot_data.dropna(subset=["GEST_GROUP", "GEMELAR_LABEL", "NICU"])
    .groupby(["GEST_GROUP", "GEMELAR_LABEL"], observed=True)["NICU"]
    .mean()
    .mul(100)
    .reset_index()
    .pivot(index="GEST_GROUP", columns="GEMELAR_LABEL", values="NICU")
)

heat = heat.reindex([idx for idx in ordem_gest if idx in heat.index])
ordem_colunas_heat = ["Gestação gemelar", "Gestação única"]
heat = heat[[col for col in ordem_colunas_heat if col in heat.columns]]

grafico_heatmap_manchester(
    heat,
    titulo=GRAFICOS[8][1],
    arquivo="grafico_09_utin_grupo_gestacional_tipo_gestacao.png",
)

# ------------------------------------------------------------
# Gráfico 10 — Taxa de UTIN segundo grupo gestacional
# ------------------------------------------------------------
tab_g10 = tabela_distribuicao(plot_data, "GEST_GROUP")
tab_g10["GEST_GROUP"] = pd.Categorical(tab_g10["GEST_GROUP"], categories=ordem_gest, ordered=True)
tab_g10 = tab_g10.sort_values("GEST_GROUP")
grafico_barras_taxa(
    tab_g10,
    "GEST_GROUP",
    titulo=GRAFICOS[9][1],
    arquivo="grafico_10_utin_grupo_gestacional.png",
    rotacao=25,
    xlabel="Grupo gestacional",
)

# ------------------------------------------------------------
# Gráfico 11 — Probabilidade de internação em UTIN por idade gestacional e pluralidade
# ------------------------------------------------------------
# Reprodução adaptada do código solicitado, agora sem restringir a idade gestacional.
# Objetivo:
# - usar todos os registros disponíveis, inclusive semanas fora da faixa 24–40;
# - manter as cores originais do ciclo padrão do Matplotlib;
# - calcular a probabilidade observada de UTIN por média da variável binária `NICU`;
# - preservar o título solicitado com a terminologia UTIN.

def plural_cat_g11(x):
    """Mapeia DPLURAL para categorias de exibição sem excluir valores extremos."""
    if pd.isna(x):
        return np.nan
    try:
        x_int = int(x)
    except Exception:
        return "Pluralidade não classificada"

    if x_int == 1:
        return "Gestação única"
    if x_int == 2:
        return "Gestação gemelar"
    if x_int == 3:
        return "Gestação trigemelar"
    if x_int == 4:
        return "Gestação quadrigemelar"
    if x_int >= 5:
        return "Gestação quíntupla ou mais"
    return "Pluralidade não classificada"

plot_data["PLURAL_CAT"] = plot_data["DPLURAL"].apply(plural_cat_g11)

# Mantém todos os valores de COMBGEST informados, inclusive outliers.
# Não há filtro por intervalo de semanas; apenas registros sem COMBGEST, PLURAL_CAT ou NICU são removidos.
df_g11 = (
    plot_data
    .dropna(subset=["COMBGEST", "PLURAL_CAT", "NICU"])
    .copy()
)

# Garante ordenação numérica da idade gestacional, caso a coluna venha como texto.
df_g11["COMBGEST"] = pd.to_numeric(df_g11["COMBGEST"], errors="coerce")
df_g11 = df_g11.dropna(subset=["COMBGEST"])

risk_g11 = (
    df_g11
    .groupby(["COMBGEST", "PLURAL_CAT"], observed=True)["NICU"]
    .agg(probabilidade="mean", n="size")
    .reset_index()
)

ordem_g11 = [
    "Gestação única",
    "Gestação gemelar",
    "Gestação trigemelar",
    "Gestação quadrigemelar",
    "Gestação quíntupla ou mais",
    "Pluralidade não classificada",
]
risk_g11["PLURAL_CAT"] = pd.Categorical(
    risk_g11["PLURAL_CAT"],
    categories=ordem_g11,
    ordered=True,
)
risk_g11 = risk_g11.sort_values(["PLURAL_CAT", "COMBGEST"])

fig, ax = plt.subplots(figsize=(12, 6), facecolor=STYLE["white"])

for p in risk_g11["PLURAL_CAT"].dropna().unique():
    subset = risk_g11[risk_g11["PLURAL_CAT"] == p]
    ax.plot(
        subset["COMBGEST"],
        subset["probabilidade"],
        marker="o",
        linewidth=1.8,
        label=str(p),
    )

apply_academic_style(
    ax,
    title="Probabilidade de internação em UTIN por idade gestacional e pluralidade",
    xlabel="Idade gestacional (semanas)",
    ylabel="Probabilidade de internação em UTIN",
    rotate_x=0,
)
ax.yaxis.set_major_formatter(PercentFormatter(xmax=1))
ax.legend(title="Tipo de gestação", frameon=False)

# Observação metodológica: semanas extremas podem ter baixo n, resultando em maior oscilação visual.
# A tabela auxiliar abaixo permite auditar o tamanho amostral por ponto do gráfico.
tabela_g11_outliers = risk_g11.rename(columns={
    "COMBGEST": "idade_gestacional_semanas",
    "PLURAL_CAT": "pluralidade",
    "probabilidade": "probabilidade_utin",
    "n": "n_registros",
})
# Salva a tabela auxiliar diretamente em Excel.
# Não usar salvar_tabela() aqui, pois essa função do pipeline espera:
# salvar_tabela(numero, titulo, df) e gera arquivo CSV numerado.
caminho_tabela_g11_outliers = (
    DIR_TABELAS / "grafico_11_dados_prob_utin_idade_gestacional_pluralidade_com_outliers.xlsx"
)
tabela_g11_outliers.to_excel(caminho_tabela_g11_outliers, index=False)
print(f"Tabela auxiliar do Gráfico 11 salva em: {caminho_tabela_g11_outliers}")

# Remove arquivos antigos do Gráfico 11, caso existam em uma execução anterior.
for arquivo_antigo_g11 in [
    "grafico_11_probabilidade_nicu_idade_gestacional_pluralidade_codigo_solicitado.png",
    "grafico_11_probabilidade_nicu_idade_gestacional_pluralidade.png",
]:
    caminho_antigo_g11 = DIR_GRAFICOS / arquivo_antigo_g11
    if caminho_antigo_g11.exists():
        caminho_antigo_g11.unlink()

salvar_figura(fig, "grafico_11_prob_utin_idade_gestacional_pluralidade.png")

# ------------------------------------------------------------
# Gráfico 12 — Distribuição do peso ao nascer segundo internação em UTIN
# ------------------------------------------------------------
grafico_boxplot_peso_utin(
    plot_data,
    titulo=GRAFICOS[11][1],
    arquivo="grafico_12_peso_nascer_utin.png",
)

# ------------------------------------------------------------
# Gráfico 13 — Taxa de UTIN segundo definição da proxy operacional
# ------------------------------------------------------------
tab_g13_long = pd.DataFrame([
    ["Conservadora", "Ausente", plot_data.loc[plot_data["PROM_PROXY_CONSERV"] == 0, "NICU"].mean() * 100],
    ["Conservadora", "Presente", plot_data.loc[plot_data["PROM_PROXY_CONSERV"] == 1, "NICU"].mean() * 100],
    ["Ampla", "Ausente", plot_data.loc[plot_data["PROM_PROXY_WIDE"] == 0, "NICU"].mean() * 100],
    ["Ampla", "Presente", plot_data.loc[plot_data["PROM_PROXY_WIDE"] == 1, "NICU"].mean() * 100],
], columns=["Definição", "Proxy", "Taxa"])

tab_g13 = (
    tab_g13_long
    .pivot(index="Definição", columns="Proxy", values="Taxa")
    .reset_index()
)

tab_g13["Definição"] = pd.Categorical(tab_g13["Definição"], categories=["Conservadora", "Ampla"], ordered=True)
tab_g13 = tab_g13.sort_values("Definição")

grafico_barras_agrupadas(
    tab_g13,
    group_col="Definição",
    value_cols=["Ausente", "Presente"],
    labels=["Proxy ausente", "Proxy presente"],
    titulo=GRAFICOS[12][1],
    arquivo="grafico_13_utin_definicao_proxy.png",
    xlabel="Definição operacional",
    ylabel="Taxa de internação em UTIN (%)",
    rotacao=0,
)

print("Gráficos 1 a 13 gerados com padrão visual harmonizado ao notebook 08.")
````

## 8. Regressões logísticas: Tabelas 22 a 28

As regressões logísticas são utilizadas como etapa interpretável de associação estatística. Elas permitem estimar razões de chances para variáveis selecionadas e avaliar a consistência dos sinais observados na análise exploratória.

Esses modelos não substituem a modelagem supervisionada final, mas oferecem uma camada explicativa complementar, especialmente para discutir prematuridade, gemelaridade, proxy operacional e variáveis maternas/obstétricas associadas à internação em UTIN.

````python
def preparar_binarias_regressao(df):
    out = df.copy()
    for c in ["RF_GDIAB", "RF_GHYPE", "WIC"]:
        out[f"{c}_BIN"] = out[c]
    return out

data_reg = preparar_binarias_regressao(data)

def ajustar_logit_or(formula, df_modelo, nome):
    """
    Ajusta regressão logística com statsmodels e retorna OR, IC95% e p-valor.

    A função é deliberadamente defensiva: em bases grandes ou desfechos compostos,
    pode ocorrer instabilidade numérica, separação quase perfeita ou intervalos muito
    amplos. Nesses casos, a tabela recebe uma coluna de observação metodológica para
    orientar a interpretação no relatório.
    """
    if not STATSMODELS_AVAILABLE:
        return pd.DataFrame({"Variável no modelo": ["erro"], "observacao_metodologica": ["statsmodels indisponível"]})

    cols = re.findall(r"[A-Za-z_][A-Za-z0-9_]*", formula)
    cols = [c for c in set(cols) if c in df_modelo.columns]
    d = df_modelo[cols].dropna().copy()

    if d.empty:
        return pd.DataFrame({
            "Variável no modelo": ["erro"],
            "observacao_metodologica": [f"{nome}: sem registros completos para a fórmula especificada"]
        })

    try:
        modelo = smf.logit(formula, data=d).fit(disp=0, maxiter=200)
        conf = modelo.conf_int()
        tab = pd.DataFrame({
            "Variável no modelo": modelo.params.index,
            "OR": np.exp(modelo.params.values),
            "IC95_inf": np.exp(conf[0].values),
            "IC95_sup": np.exp(conf[1].values),
            "p_valor": modelo.pvalues.values,
        })

        tab["n_registros_modelo"] = len(d)
        tab["convergiu"] = bool(getattr(modelo, "mle_retvals", {}).get("converged", True))

        # Sinaliza ORs/ICs potencialmente instáveis, sem excluir automaticamente a saída.
        cond_extremo = (
            (~np.isfinite(tab[["OR", "IC95_inf", "IC95_sup"]]).all(axis=1))
            | (tab["OR"].abs() > 1e6)
            | (tab["IC95_sup"].abs() > 1e8)
            | ((tab["IC95_sup"] / tab["IC95_inf"].replace(0, np.nan)).abs() > 1e8)
        )
        tab["observacao_metodologica"] = np.where(
            cond_extremo,
            "Interpretar com cautela: possível separação, circularidade ou instabilidade numérica.",
            "OK"
        )
        if not bool(tab["convergiu"].all()):
            tab["observacao_metodologica"] = np.where(
                tab["observacao_metodologica"].eq("OK"),
                "Interpretar com cautela: modelo não convergiu plenamente.",
                tab["observacao_metodologica"] + " Modelo não convergiu plenamente."
            )
        return tab

    except Exception as e:
        return pd.DataFrame({
            "Variável no modelo": ["erro"],
            "OR": [np.nan],
            "IC95_inf": [np.nan],
            "IC95_sup": [np.nan],
            "p_valor": [np.nan],
            "n_registros_modelo": [len(d)],
            "convergiu": [False],
            "observacao_metodologica": [f"{nome}: regressão não estimada com segurança ({type(e).__name__}: {e})"]
        })

def extrair_linha_modelo(tab, variavel, proxy=None, modelo=None):
    """Extrai uma linha de resultado sem quebrar o pipeline caso a regressão falhe."""
    if "Variável no modelo" not in tab.columns:
        row = {"Variável no modelo": variavel, "OR": np.nan, "IC95_inf": np.nan, "IC95_sup": np.nan, "p_valor": np.nan,
               "observacao_metodologica": "Tabela sem coluna de variável; resultado indisponível."}
    else:
        sel = tab[tab["Variável no modelo"] == variavel]
        if sel.empty:
            row = {"Variável no modelo": variavel, "OR": np.nan, "IC95_inf": np.nan, "IC95_sup": np.nan, "p_valor": np.nan,
                   "observacao_metodologica": "Variável não estimada ou removida por instabilidade/ausência de dados completos."}
        else:
            row = sel.iloc[0].to_dict()
    if proxy is not None:
        row["Proxy"] = proxy
    if modelo is not None:
        row["Modelo"] = modelo
    return row

formula_22 = "NICU ~ PROM_PROXY_WIDE * GEMELAR + MAGER + RF_GDIAB_BIN + RF_GHYPE_BIN + PREVIS + WIC_BIN"
formula_23 = formula_22 + " + COMBGEST + DBWT"

tabela_22 = ajustar_logit_or(formula_22, data_reg, "tabela_22")
tabela_23 = ajustar_logit_or(formula_23, data_reg, "tabela_23")
salvar_tabela(22, TABELAS[21][1], tabela_22)
salvar_tabela(23, TABELAS[22][1], tabela_23)

# Tabela 24: sensibilidade de proxies
linhas = []
for proxy in ["PROM_PROXY_WIDE", "PROM_PROXY_CONSERV"]:
    for nome_modelo, formula in [
        ("Associação total", f"NICU ~ {proxy} + GEMELAR + MAGER + RF_GDIAB_BIN + RF_GHYPE_BIN + PREVIS + WIC_BIN"),
        ("Ajustado por COMBGEST e DBWT", f"NICU ~ {proxy} + GEMELAR + MAGER + RF_GDIAB_BIN + RF_GHYPE_BIN + PREVIS + WIC_BIN + COMBGEST + DBWT")
    ]:
        tab = ajustar_logit_or(formula, data_reg, nome_modelo)
        linhas.append(extrair_linha_modelo(tab, proxy, proxy=proxy, modelo=nome_modelo))
tabela_24 = pd.DataFrame(linhas)
salvar_tabela(24, TABELAS[23][1], tabela_24)

# Tabela 25: associação descritiva entre proxy ampla, gemelaridade e internação em UTIN.
tabela_25 = (
    data.dropna(subset=["GEMELAR", "PROM_PROXY_WIDE", "NICU"])
    .groupby(["GEMELAR", "PROM_PROXY_WIDE"], observed=True)["NICU"]
    .agg(total_registros="count", casos_utin="sum", taxa_utin="mean")
    .reset_index()
)
tabela_25["taxa_utin_pct"] = tabela_25["taxa_utin"] * 100
tabela_25["gemelaridade"] = tabela_25["GEMELAR"].map({0: "Não gemelar", 1: "Gemelar"})
tabela_25["proxy_operacional_ampla"] = tabela_25["PROM_PROXY_WIDE"].map({0: "Proxy ampla negativa", 1: "Proxy ampla positiva"})
tabela_25 = tabela_25[[
    "gemelaridade", "proxy_operacional_ampla", "GEMELAR", "PROM_PROXY_WIDE",
    "total_registros", "casos_utin", "taxa_utin", "taxa_utin_pct"
]]
salvar_tabela(25, TABELAS[24][1], tabela_25)

# Tabelas 26 e 27: desfecho neonatal grave
formula_26 = "DESFECHO_GRAVE_SEM_COMPONENTES ~ PROM_PROXY_WIDE * GEMELAR + MAGER + RF_GDIAB_BIN + RF_GHYPE_BIN + PREVIS + WIC_BIN + LD_CHOR"
formula_27 = formula_26 + " + COMBGEST + DBWT"
tabela_26 = ajustar_logit_or(formula_26, data_reg, "tabela_26")
tabela_27 = ajustar_logit_or(formula_27, data_reg, "tabela_27")
salvar_tabela(26, TABELAS[25][1], tabela_26)
salvar_tabela(27, TABELAS[26][1], tabela_27)

# Tabela 28: métrica simples da regressão logística segura
features_seguras = ["MAGER", "PREVIS", "WIC", "RF_GDIAB", "RF_GHYPE", "COMBGEST", "DBWT", "DPLURAL", "GEMELAR", "LD_CHOR", "LD_ANTB", "LD_STER", "LD_INDL", "PROM_PROXY_WIDE"]
model_df = data[features_seguras + ["NICU"]].dropna(subset=["NICU"])
X = model_df[features_seguras]
y = model_df["NICU"].astype(int)
imp = SimpleImputer(strategy="median")
X_imp = imp.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_imp, y, test_size=0.25, stratify=y, random_state=RANDOM_STATE)
logit_pipe = Pipeline([("scaler", StandardScaler()), ("model", LogisticRegression(max_iter=5000, class_weight="balanced"))])
logit_pipe.fit(X_train, y_train)
prob = logit_pipe.predict_proba(X_test)[:, 1]
pred = (prob >= 0.5).astype(int)
tabela_28 = pd.DataFrame([{
    "modelo": "Regressão logística segura",
    "AUC_ROC": roc_auc_score(y_test, prob),
    "PR_AUC": average_precision_score(y_test, prob),
    "acuracia": accuracy_score(y_test, pred),
    "precisao": precision_score(y_test, pred, zero_division=0),
    "sensibilidade": recall_score(y_test, pred, zero_division=0),
    "F1": f1_score(y_test, pred, zero_division=0),
    "Brier": brier_score_loss(y_test, prob)
}])
salvar_tabela(28, TABELAS[27][1], tabela_28)
print("Tabelas 22 a 28 geradas.")
````

## 9. Modelagem supervisionada: cenários completo/perinatal e restritivo

Esta seção treina e avalia modelos supervisionados em dois cenários analíticos complementares. A finalidade metodológica é **estratificar retrospectivamente o risco observado de internação em UTIN** em registros populacionais consolidados, e não produzir uma ferramenta clínica prospectiva.

O **cenário completo/perinatal** estima o desempenho retrospectivo máximo em registros consolidados, incluindo variáveis maternas, obstétricas, perinatais e neonatais disponíveis na base. Por conter informações próximas à gravidade neonatal imediata, como Apgar, ventilação, surfactante, antibiótico neonatal e convulsões, esse cenário exige cautela interpretativa quanto ao risco de vazamento temporal de informação. Ele responde à pergunta: *quão bem os registros consolidados permitem organizar retrospectivamente os nascimentos por risco observado de UTIN?*

O **cenário restritivo final** remove variáveis neonatais diretas e variáveis com maior risco de disponibilidade tardia ou proximidade direta com o desfecho assistencial. Seu objetivo é oferecer uma leitura mais conservadora da capacidade discriminativa, mesmo com redução esperada de desempenho. Ainda assim, esse cenário mantém idade gestacional (`COMBGEST`) e peso ao nascer (`DBWT`), variáveis fortemente relacionadas à vulnerabilidade neonatal. Portanto, ele **reduz, mas não elimina completamente**, a proximidade temporal entre preditores e desfecho.

A comparação entre os cenários é central para o TCC, pois demonstra que a qualidade do modelo não deve ser avaliada apenas por AUC-ROC. Também devem ser considerados disponibilidade temporal dos preditores, calibração, limiares de decisão, risco de vazamento, validade externa e finalidade de uso.

A regressão logística é interpretada como associação observacional; XGBoost e Random Forest são interpretados como modelos classificatórios; SHAP e medidas de importância indicam contribuição algorítmica, não causalidade.

````python
# Garantia para execuções parciais: define a aceleração do XGBoost caso a célula de configuração não tenha sido executada.
if "PARAMETROS_XGBOOST_ACELERACAO" not in globals():
    def detectar_cuda_nvidia_smi():
        try:
            resultado = subprocess.run(
                ["nvidia-smi"],
                stdout=subprocess.PIPE,
                stderr=subprocess.PIPE,
                text=True,
                timeout=5,
            )
            return resultado.returncode == 0
        except Exception:
            return False

    def _versao_xgboost_major():
        if not XGBOOST_AVAILABLE or xgb is None:
            return None
        try:
            return int(str(xgb.__version__).split(".")[0])
        except Exception:
            return None

    def configurar_aceleracao_xgboost():
        modo = os.environ.get("TCC_XGBOOST_DEVICE", "auto").strip().lower()
        if modo not in {"auto", "gpu", "cuda", "cpu"}:
            modo = "auto"
        cuda_detectado = detectar_cuda_nvidia_smi()
        major = _versao_xgboost_major()
        solicitar_gpu = modo in {"auto", "gpu", "cuda"} and cuda_detectado and XGBOOST_AVAILABLE
        if solicitar_gpu and major is not None and major >= 2:
            parametros = {"tree_method": "hist", "device": "cuda"}
            usando_gpu = True
            estrategia = 'XGBoost 2.x: tree_method="hist", device="cuda"'
        elif solicitar_gpu:
            parametros = {"tree_method": "gpu_hist", "predictor": "gpu_predictor"}
            usando_gpu = True
            estrategia = 'XGBoost 1.x: tree_method="gpu_hist", predictor="gpu_predictor"'
        else:
            parametros = {"tree_method": "hist"}
            usando_gpu = False
            estrategia = 'CPU/fallback: tree_method="hist"'
        return {
            "MODO_ACELERACAO_XGBOOST": modo,
            "CUDA_DETECTADO": bool(cuda_detectado),
            "USAR_GPU_XGBOOST": bool(usando_gpu),
            "PARAMETROS_XGBOOST_ACELERACAO": parametros,
            "ESTRATEGIA_XGBOOST_ACELERACAO": estrategia,
            "XGBOOST_VERSAO_MAJOR": major,
        }

    _config_xgb_gpu = configurar_aceleracao_xgboost()
    MODO_ACELERACAO_XGBOOST = _config_xgb_gpu["MODO_ACELERACAO_XGBOOST"]
    CUDA_DETECTADO = _config_xgb_gpu["CUDA_DETECTADO"]
    USAR_GPU_XGBOOST = _config_xgb_gpu["USAR_GPU_XGBOOST"]
    PARAMETROS_XGBOOST_ACELERACAO = _config_xgb_gpu["PARAMETROS_XGBOOST_ACELERACAO"]
    ESTRATEGIA_XGBOOST_ACELERACAO = _config_xgb_gpu["ESTRATEGIA_XGBOOST_ACELERACAO"]
    XGBOOST_VERSAO_MAJOR = _config_xgb_gpu["XGBOOST_VERSAO_MAJOR"]

# Cenário completo/perinatal: inclui variáveis próximas da gravidade neonatal imediata.
# Útil para estimar capacidade retrospectiva máxima, mas com maior risco de vazamento temporal.
features_completo = [
    "MAGER", "PREVIS", "WIC", "RF_GDIAB", "RF_GHYPE",
    "COMBGEST", "DBWT", "DPLURAL", "GEMELAR",
    "LD_CHOR", "LD_ANTB", "LD_STER", "LD_INDL",
    "APGAR5", "AB_AVEN1", "AB_AVEN6", "AB_SURF", "AB_ANTI", "AB_SEIZ",
    "PROM_PROXY_WIDE"
]

# Cenário restritivo final: remove variáveis neonatais diretas do cenário completo.
# Mantém COMBGEST e DBWT; portanto, reduz, mas não elimina, a proximidade temporal com o desfecho.
features_restritivo = [
    "MAGER", "PREVIS", "WIC", "RF_GDIAB", "RF_GHYPE",
    "COMBGEST", "DBWT", "DPLURAL", "GEMELAR",
    "LD_CHOR", "LD_ANTB", "LD_STER", "LD_INDL",
    "PROM_PROXY_WIDE"
]

# Limiares decisórios definidos para reproduzir as tabelas finais do documento.
# O modelo completo/perinatal prioriza sensibilidade retrospectiva; o restritivo usa limiar mais conservador.
LIMIARES_XGBOOST = {
    "completo_perinatal": 0.20,
    "restritivo_final": 0.25,
}

# O documento de referência usa conjunto de teste final com aproximadamente 20% da base
# (matrizes finais com 726.271 registros). Manter este valor fixo evita divergência
# entre as Tabelas 43, 44 e 45 do notebook e do texto.
TEST_SIZE_XGBOOST_FINAL = 0.20

# Guardrails metodológicos dos cenários.
# Estes metadados são exportados para documentar a finalidade analítica e evitar leitura clínica indevida.
METODOLOGIA_MODELOS = {
    "completo_perinatal": {
        "finalidade_analitica": "Estratificação retrospectiva máxima em registros consolidados.",
        "pergunta_analitica": "Quão bem variáveis maternas, obstétricas, perinatais e neonatais organizam retrospectivamente o risco observado de UTIN?",
        "risco_vazamento_temporal": "Alto/moderado: inclui variáveis próximas da gravidade neonatal imediata.",
        "uso_clinico_prospectivo": "Não validado; não usar para tomada de decisão clínica individual.",
        "interpretacao_recomendada": "Resultado de Health Analytics retrospectivo, útil para análise populacional e discussão metodológica."
    },
    "restritivo_final": {
        "finalidade_analitica": "Estratificação retrospectiva conservadora com remoção de variáveis neonatais diretas.",
        "pergunta_analitica": "Quanto desempenho permanece após reduzir variáveis temporalmente próximas do desfecho?",
        "risco_vazamento_temporal": "Reduzido, mas não nulo: mantém idade gestacional e peso ao nascer.",
        "uso_clinico_prospectivo": "Não validado; não interpretar como modelo pré-natal ou assistencial prospectivo.",
        "interpretacao_recomendada": "Cenário mais prudente para comparar desempenho e discutir robustez metodológica."
    },
    "desfecho_neonatal_grave": {
        "finalidade_analitica": "Análise complementar de desfecho composto grave.",
        "pergunta_analitica": "Como o modelo se comporta em um desfecho neonatal composto?",
        "risco_vazamento_temporal": "Depende dos componentes do desfecho; interpretar como sensibilidade retrospectiva.",
        "uso_clinico_prospectivo": "Não validado; análise complementar.",
        "interpretacao_recomendada": "Sensibilidade metodológica, não substitui a análise principal de UTIN."
    }
}

pd.DataFrame.from_dict(METODOLOGIA_MODELOS, orient="index").reset_index(names="cenario").to_csv(
    DIR_QUADROS / "quadro_aux_guardrails_cenarios_modelagem.csv",
    index=False,
    encoding="utf-8-sig"
)

PROXY_RPM_PPROM_DOCUMENTACAO = {
    "variavel": "PROM_PROXY_WIDE",
    "tipo": "proxy operacional retrospectiva",
    "equivalencia_diagnostica": "Não possui equivalência diagnóstica individual com RPM/PPROM.",
    "uso_permitido": "Classificação analítica de registros compatíveis com vulnerabilidade pré-termo e marcadores obstétricos/assistenciais disponíveis.",
    "uso_nao_permitido": "Não estimar prevalência clínica real de RPM/PPROM; não afirmar causalidade; não usar como diagnóstico.",
    "alerta_circularidade": "A proxy inclui prematuridade e deve ser interpretada junto ao risco de circularidade com UTIN."
}
pd.DataFrame([PROXY_RPM_PPROM_DOCUMENTACAO]).to_csv(
    DIR_QUADROS / "quadro_aux_documentacao_proxy_rpm_pprom.csv",
    index=False,
    encoding="utf-8-sig"
)


variaveis_removidas_vazamento = sorted(set(features_completo) - set(features_restritivo))
pd.DataFrame({"variavel_removida": variaveis_removidas_vazamento}).to_csv(
    DIR_QUADROS / "quadro_29_variaveis_removidas_por_vazamento.csv", index=False, encoding="utf-8-sig"
)


def preparar_xy(df, features, target="NICU"):
    """
    Prepara X/y de forma econômica em memória.

    Correção aplicada:
    - evita SimpleImputer(strategy="median").fit_transform(X) na base inteira;
    - converte preditores para float32;
    - deixa a imputação para depois da divisão treino/teste, usando apenas medianas do treino.
    Isso reduz o uso de memória e evita MemoryError em bases com milhões de registros.
    """
    features = [c for c in features if c in df.columns]
    if target not in df.columns:
        raise KeyError(f"Variável-alvo ausente no DataFrame: {target}")

    mask = df[target].notna()
    idx = df.index[mask]
    X = df.loc[mask, features].apply(pd.to_numeric, errors="coerce").astype(np.float32)
    y = df.loc[mask, target].astype(np.int8)
    return X, y, features, idx

def imputar_por_mediana_treino(X_train_raw, X_test_raw):
    """
    Imputa valores ausentes com medianas calculadas somente no treino.
    Usa pandas/float32 para evitar a cópia densa grande criada pelo SimpleImputer.
    """
    medianas = X_train_raw.median(axis=0, numeric_only=True).fillna(0).astype(np.float32)
    X_train = X_train_raw.fillna(medianas).astype(np.float32)
    X_test = X_test_raw.fillna(medianas).astype(np.float32)
    return X_train, X_test, medianas

def criar_xgb(scale_pos_weight=None):
    if not XGBOOST_AVAILABLE:
        return None
    kwargs = dict(
        n_estimators=400,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=RANDOM_STATE,
    )
    kwargs.update(PARAMETROS_XGBOOST_ACELERACAO)
    if scale_pos_weight is not None:
        kwargs["scale_pos_weight"] = scale_pos_weight
    return XGBClassifier(**kwargs)

def calibrar_modelo_prefit(modelo_base, X_cal, y_cal, metodo="isotonic"):
    """
    Calibra um modelo já treinado usando um subconjunto de calibração.

    Esta versão substitui CalibratedClassifierCV(..., cv=3), que treinava três clones
    do XGBoost e elevava muito o consumo de memória. Primeiro tenta a API nova
    FrozenEstimator; se indisponível, usa cv="prefit"; se ainda falhar, retorna o
    modelo base sem calibração para não interromper o pipeline.
    """
    try:
        from sklearn.frozen import FrozenEstimator
        calibrado = CalibratedClassifierCV(FrozenEstimator(modelo_base), method=metodo)
        calibrado.fit(X_cal, y_cal)
        return calibrado, f"CalibratedClassifierCV com FrozenEstimator, método={metodo}"
    except Exception:
        try:
            calibrado = CalibratedClassifierCV(modelo_base, method=metodo, cv="prefit")
            calibrado.fit(X_cal, y_cal)
            return calibrado, f"CalibratedClassifierCV cv='prefit', método={metodo}"
        except Exception as e:
            print(f"Aviso: calibração não executada ({type(e).__name__}: {e}). Usando probabilidades do XGBoost base.")
            return modelo_base, "Sem calibração; fallback para modelo base"

def avaliar_prob(y_true, prob, nome, limiar=0.50):
    """Avalia probabilidades e métricas classificatórias em limiar explícito."""
    pred = (prob >= limiar).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, pred).ravel()
    especificidade = tn / (tn + fp) if (tn + fp) else np.nan
    return {
        "modelo": nome,
        "limiar_classificacao": float(limiar),
        "AUC_ROC": roc_auc_score(y_true, prob),
        "PR_AUC": average_precision_score(y_true, prob),
        "acuracia": accuracy_score(y_true, pred),
        "precisao": precision_score(y_true, pred, zero_division=0),
        "sensibilidade": recall_score(y_true, pred, zero_division=0),
        "especificidade": especificidade,
        "F1": f1_score(y_true, pred, zero_division=0),
        "Brier": brier_score_loss(y_true, prob)
    }

def treinar_xgb_cenario(df, features, target, nome_cenario):
    """
    Treina XGBoost com divisão treino/calibração/teste e menor uso de memória.

    Etapas:
    1. separa teste final estratificado;
    2. calcula medianas apenas no treino e imputa treino/teste;
    3. separa uma fração do treino para calibração;
    4. treina XGBoost uma única vez;
    5. calibra o modelo pré-ajustado, evitando cv=3 e clones pesados.
    """
    import gc

    X_raw, y, feats, idx = preparar_xy(df, features, target)

    X_train_raw, X_test_raw, y_train, y_test, idx_train, idx_test = train_test_split(
        X_raw, y, idx, test_size=TEST_SIZE_XGBOOST_FINAL, stratify=y, random_state=RANDOM_STATE
    )

    # Libera referências grandes assim que possível.
    del X_raw
    gc.collect()

    X_train, X_test, medianas_imputacao = imputar_por_mediana_treino(X_train_raw, X_test_raw)
    del X_train_raw, X_test_raw
    gc.collect()

    # Subconjunto de calibração separado do treino do modelo.
    X_model, X_cal, y_model, y_cal = train_test_split(
        X_train, y_train, test_size=0.20, stratify=y_train, random_state=RANDOM_STATE
    )
    del X_train, y_train
    gc.collect()

    ratio = (y_model == 0).sum() / max((y_model == 1).sum(), 1)
    modelo_base = criar_xgb(scale_pos_weight=ratio)
    if modelo_base is None:
        raise ImportError("XGBoost indisponível.")

    modelo_base.fit(X_model, y_model)
    del X_model, y_model
    gc.collect()

    # Probabilidades originais do XGBoost antes da calibração.
    # Guardadas para permitir os gráficos comparativos "Sem calibração" x "Calibrado".
    prob_sem_calibracao = modelo_base.predict_proba(X_test)[:, 1]

    modelo_calibrado, metodo_calibracao = calibrar_modelo_prefit(modelo_base, X_cal, y_cal, metodo="isotonic")
    del X_cal, y_cal
    gc.collect()

    prob = modelo_calibrado.predict_proba(X_test)[:, 1]
    limiar = LIMIARES_XGBOOST.get(nome_cenario, 0.50)
    metrics = avaliar_prob(y_test, prob, f"XGBoost calibrado — {nome_cenario}", limiar=limiar)
    guardrail = METODOLOGIA_MODELOS.get(nome_cenario, {})
    metrics["metodo_calibracao"] = metodo_calibracao
    metrics["n_teste"] = int(len(y_test))
    metrics["n_features"] = int(len(feats))
    metrics["finalidade_analitica"] = guardrail.get("finalidade_analitica", "Estratificação retrospectiva em registros consolidados.")
    metrics["risco_vazamento_temporal"] = guardrail.get("risco_vazamento_temporal", "Avaliar conforme variáveis disponíveis.")
    metrics["uso_clinico_prospectivo"] = guardrail.get("uso_clinico_prospectivo", "Não validado para uso clínico prospectivo.")
    metrics["interpretacao_recomendada"] = guardrail.get("interpretacao_recomendada", "Resultado retrospectivo; não causal.")

    return {
        "nome": nome_cenario,
        "modelo_base": modelo_base,
        "modelo_calibrado": modelo_calibrado,
        "X_test": X_test,
        "y_test": y_test,
        "idx_test": idx_test,
        "prob": prob,
        "prob_sem_calibracao": prob_sem_calibracao,
        "features": feats,
        "metrics": metrics,
        "limiar": limiar,
        "medianas_imputacao": medianas_imputacao,
        "metodo_calibracao": metodo_calibracao,
    }

print("Aceleração efetiva do XGBoost:", PARAMETROS_XGBOOST_ACELERACAO, "|", globals().get("ESTRATEGIA_XGBOOST_ACELERACAO", ""))
resultados_xgb = {}
if XGBOOST_AVAILABLE:
    resultados_xgb["completo"] = treinar_xgb_cenario(data, features_completo, "NICU", "completo_perinatal")
    resultados_xgb["restritivo"] = treinar_xgb_cenario(data, features_restritivo, "NICU", "restritivo_final")
else:
    print("XGBoost indisponível; blocos de XGBoost não serão executados.")
````

````python
# ============================================================
# Guardrails dos cenários de modelagem e risco de vazamento temporal
# ============================================================

def classificar_variaveis_por_tempo(features):
    proximas_desfecho = {"APGAR5", "AB_AVEN1", "AB_AVEN6", "AB_SURF", "AB_ANTI", "AB_SEIZ"}
    perinatais_intermediarias = {"COMBGEST", "DBWT", "DPLURAL", "GEMELAR"}
    obstetricas_maternas = {"MAGER", "PREVIS", "WIC", "RF_GDIAB", "RF_GHYPE", "LD_CHOR", "LD_ANTB", "LD_STER", "LD_INDL", "PROM_PROXY_WIDE"}
    linhas = []
    for f in features:
        if f in proximas_desfecho:
            grupo = "Neonatal/assistencial próxima do desfecho"
            risco = "Maior risco de vazamento temporal"
        elif f in perinatais_intermediarias:
            grupo = "Perinatal/intermediária"
            risco = "Risco intermediário; exige interpretação retrospectiva"
        elif f in obstetricas_maternas:
            grupo = "Materna/obstétrica/proxy operacional"
            risco = "Menor risco relativo, mas ainda observacional"
        else:
            grupo = "Não classificada"
            risco = "Revisar"
        linhas.append({"Variável": f, "Grupo temporal/metodológico": grupo, "Leitura recomendada": risco})
    return pd.DataFrame(linhas)

matriz_temporal_completo = classificar_variaveis_por_tempo(features_completo)
matriz_temporal_restritivo = classificar_variaveis_por_tempo(features_restritivo)

SINTESE_CENARIOS_MODELAGEM = pd.DataFrame([
    {
        "Cenário": "Completo/perinatal",
        "Número de variáveis": len(features_completo),
        "Finalidade": "Estimar capacidade retrospectiva máxima de organização do risco em registros consolidados.",
        "Cautela": "Inclui variáveis neonatais/assistenciais próximas ao desfecho; não deve ser lido como modelo prospectivo."
    },
    {
        "Cenário": "Restritivo final",
        "Número de variáveis": len(features_restritivo),
        "Finalidade": "Oferecer avaliação mais conservadora, reduzindo variáveis mais próximas do desfecho.",
        "Cautela": "Reduz, mas não elimina totalmente, proximidade temporal; mantém variáveis perinatais como COMBGEST e DBWT."
    },
])

try:
    SINTESE_CENARIOS_MODELAGEM.to_csv(DIR_QUADROS / "quadro_07_comparacao_interpretativa_cenarios_modelagem.csv", index=False, encoding="utf-8-sig")
    matriz_temporal_completo.to_csv(DIR_QUADROS / "quadro_matriz_temporal_variaveis_completo_perinatal.csv", index=False, encoding="utf-8-sig")
    matriz_temporal_restritivo.to_csv(DIR_QUADROS / "quadro_matriz_temporal_variaveis_restritivo_final.csv", index=False, encoding="utf-8-sig")
except Exception as exc:
    print("Aviso: exportação da matriz temporal não realizada:", exc)

display(Markdown("### Síntese interpretativa dos cenários"))
display(SINTESE_CENARIOS_MODELAGEM)

display(Markdown("### Matriz temporal — cenário completo/perinatal"))
display(matriz_temporal_completo)

display(Markdown("### Matriz temporal — cenário restritivo final"))
display(matriz_temporal_restritivo)
````

## 10. Tabelas 29 a 45 e Gráficos 14 a 21: avaliação dos modelos

Esta seção consolida as principais saídas de modelagem do TCC: relatórios de classificação, curvas ROC, curvas precisão-sensibilidade, calibração probabilística, concentração de risco, importância de variáveis, contribuições nativas do XGBoost e comparação final dos cenários.

A calibração é avaliada como ajuste da **qualidade probabilística**, não como tentativa de aumentar artificialmente a AUC. Portanto, uma AUC-ROC semelhante antes e depois da calibração não indica erro: a AUC mede principalmente a ordenação dos casos, enquanto a calibração avalia se as probabilidades previstas se aproximam das frequências observadas.

Na execução 0, o notebook exporta uma tabela auxiliar de diagnóstico da calibração, comparando o modelo sem calibração e o modelo calibrado por AUC-ROC, PR-AUC, Brier Score, diferença média absoluta das probabilidades, correlação de ordenação e verificação de identidade exata das probabilidades.

As métricas e os gráficos desta seção devem ser lidos como resultados retrospectivos em dados secundários, sem validação para uso clínico prospectivo.

Observação metodológica: as métricas probabilísticas finais são calculadas com os modelos calibrados, enquanto a importância das variáveis e as contribuições tipo SHAP nativas do XGBoost são extraídas do classificador base. Assim, a explicabilidade deve ser entendida como interpretação do modelo base utilizado antes da calibração probabilística, não como inferência causal.

````python
def aplicar_legenda_padrao_notebook02(ax, loc="best", title=None):
    """
    Padroniza a legenda no estilo usado no notebook 02:
    legenda simples, objetiva, com identificação da série/modelo e métrica no próprio rótulo.
    """
    ax.legend(loc=loc, title=title, frameon=True)


def salvar_curva_roc(y, prob_calibrado, titulo, arquivo, prob_sem_calibracao=None, nome_modelo="XGBoost"):
    """
    Salva curva ROC com legenda refatorada no padrão do notebook 02:
    Nome do modelo + métrica AUC no rótulo da linha.
    """
    fig, ax = plt.subplots(figsize=(8, 6))

    if prob_sem_calibracao is not None:
        fpr_sem, tpr_sem, _ = roc_curve(y, prob_sem_calibracao)
        auc_sem = roc_auc_score(y, prob_sem_calibracao)
        ax.plot(
            fpr_sem,
            tpr_sem,
            marker="o",
            markersize=3,
            linewidth=1.8,
            label=f"{nome_modelo} sem calibração (AUC = {auc_sem:.3f})",
        )

    fpr_cal, tpr_cal, _ = roc_curve(y, prob_calibrado)
    auc_cal = roc_auc_score(y, prob_calibrado)
    ax.plot(
        fpr_cal,
        tpr_cal,
        marker="o",
        markersize=3,
        linewidth=1.8,
        label=f"{nome_modelo} calibrado (AUC = {auc_cal:.3f})",
    )

    ax.plot([0, 1], [0, 1], linestyle="--", label="Aleatório")
    ax.set_title(titulo)
    ax.set_xlabel("Taxa de falso positivo")
    ax.set_ylabel("Taxa de verdadeiro positivo")
    aplicar_legenda_padrao_notebook02(ax, loc="lower right")
    fig.tight_layout()
    fig.savefig(DIR_GRAFICOS / arquivo, dpi=300, bbox_inches="tight")
    plt.show()


def salvar_curva_pr(y, prob_calibrado, titulo, arquivo, prob_sem_calibracao=None, nome_modelo="XGBoost"):
    """
    Salva curva precisão-sensibilidade com legenda no mesmo padrão:
    Nome da série + PR-AUC no rótulo.
    """
    fig, ax = plt.subplots(figsize=(8, 6))

    if prob_sem_calibracao is not None:
        prec_sem, rec_sem, _ = precision_recall_curve(y, prob_sem_calibracao)
        pr_auc_sem = average_precision_score(y, prob_sem_calibracao)
        ax.plot(
            rec_sem,
            prec_sem,
            marker="o",
            markersize=3,
            linewidth=1.8,
            label=f"{nome_modelo} sem calibração (PR-AUC = {pr_auc_sem:.3f})",
        )

    prec_cal, rec_cal, _ = precision_recall_curve(y, prob_calibrado)
    pr_auc_cal = average_precision_score(y, prob_calibrado)
    ax.plot(
        rec_cal,
        prec_cal,
        marker="o",
        markersize=3,
        linewidth=1.8,
        label=f"{nome_modelo} calibrado (PR-AUC = {pr_auc_cal:.3f})",
    )

    prevalencia = float(np.mean(y))
    ax.hlines(
        prevalencia,
        xmin=0,
        xmax=1,
        linestyle="--",
        label=f"Linha de base (prevalência = {prevalencia:.3f})",
    )
    ax.set_title(titulo)
    ax.set_xlabel("Recall / sensibilidade")
    ax.set_ylabel("Precisão")
    aplicar_legenda_padrao_notebook02(ax, loc="best")
    fig.tight_layout()
    fig.savefig(DIR_GRAFICOS / arquivo, dpi=300, bbox_inches="tight")
    plt.show()


def salvar_calibracao(y, prob_calibrado, titulo, arquivo, prob_sem_calibracao=None, nome_modelo="XGBoost"):
    """
    Salva curva de calibração com legenda objetiva:
    séries antes/depois da calibração e linha de calibração ideal.
    """
    fig, ax = plt.subplots(figsize=(8, 6))

    if prob_sem_calibracao is not None:
        frac_sem, mean_sem = calibration_curve(y, prob_sem_calibracao, n_bins=10)
        ax.plot(
            mean_sem,
            frac_sem,
            marker="o",
            linewidth=1.8,
            label=f"{nome_modelo} sem calibração",
        )

    frac_cal, mean_cal = calibration_curve(y, prob_calibrado, n_bins=10)
    ax.plot(
        mean_cal,
        frac_cal,
        marker="o",
        linewidth=1.8,
        label=f"{nome_modelo} calibrado",
    )

    ax.plot([0, 1], [0, 1], linestyle="--", label="Calibração ideal")
    ax.set_title(titulo)
    ax.set_xlabel("Probabilidade prevista média")
    ax.set_ylabel("Frequência observada")
    aplicar_legenda_padrao_notebook02(ax, loc="upper left")
    fig.tight_layout()
    fig.savefig(DIR_GRAFICOS / arquivo, dpi=300, bbox_inches="tight")
    plt.show()


def diagnosticar_calibracao(y, prob_sem, prob_cal, nome_cenario):
    """
    Diagnóstico metodológico da calibração.

    AUC-ROC e PR-AUC avaliam ordenação/discriminação.
    Brier Score e diferença média absoluta avaliam alteração probabilística.
    A correlação de postos mostra se a calibração preservou a ordenação dos riscos.
    """
    y = pd.Series(y).reset_index(drop=True).astype(int)
    prob_sem = pd.Series(prob_sem).reset_index(drop=True).astype(float)
    prob_cal = pd.Series(prob_cal).reset_index(drop=True).astype(float)

    return pd.DataFrame([{
        "cenario": nome_cenario,
        "AUC_sem_calibracao": roc_auc_score(y, prob_sem),
        "AUC_calibrado": roc_auc_score(y, prob_cal),
        "delta_AUC_calibrado_menos_sem": roc_auc_score(y, prob_cal) - roc_auc_score(y, prob_sem),
        "PR_AUC_sem_calibracao": average_precision_score(y, prob_sem),
        "PR_AUC_calibrado": average_precision_score(y, prob_cal),
        "Brier_sem_calibracao": brier_score_loss(y, prob_sem),
        "Brier_calibrado": brier_score_loss(y, prob_cal),
        "delta_Brier_calibrado_menos_sem": brier_score_loss(y, prob_cal) - brier_score_loss(y, prob_sem),
        "diferenca_media_absoluta_prob": float(np.mean(np.abs(prob_cal - prob_sem))),
        "diferenca_maxima_absoluta_prob": float(np.max(np.abs(prob_cal - prob_sem))),
        "correlacao_spearman_postos": float(prob_sem.rank(method="average").corr(prob_cal.rank(method="average"))),
        "probabilidades_identicas": bool(np.allclose(prob_sem.values, prob_cal.values, rtol=0, atol=1e-12)),
        "interpretacao": "AUC semelhante é esperada quando a calibração preserva a ordenação; verificar Brier e diferenças probabilísticas."
    }])


def tabela_classificacao(y, prob, limiar=0.50):
    """Relatório de classificação com limiar explícito."""
    pred = (prob >= limiar).astype(int)
    report = classification_report(y, pred, output_dict=True, zero_division=0)
    out = pd.DataFrame(report).T.reset_index().rename(columns={"index": "classe"})
    out.insert(0, "limiar_classificacao", float(limiar))
    return out

def tabela_matriz_confusao(y, prob, limiar=0.50):
    """Matriz de confusão com limiar explícito."""
    pred = (prob >= limiar).astype(int)
    cm = confusion_matrix(y, pred)
    out = pd.DataFrame(cm, index=["Real 0", "Real 1"], columns=["Predito 0", "Predito 1"]).reset_index()
    out.insert(0, "limiar_classificacao", float(limiar))
    return out

if resultados_xgb:
    comp = resultados_xgb["completo"]
    rest = resultados_xgb["restritivo"]

    tabela_aux_diagnostico_calibracao = pd.concat([
        diagnosticar_calibracao(comp["y_test"], comp["prob_sem_calibracao"], comp["prob"], "completo_perinatal"),
        diagnosticar_calibracao(rest["y_test"], rest["prob_sem_calibracao"], rest["prob"], "restritivo_final"),
    ], ignore_index=True)
    tabela_aux_diagnostico_calibracao.to_csv(
        DIR_TABELAS / "tabela_aux_diagnostico_calibracao_xgb.csv",
        index=False,
        encoding="utf-8-sig"
    )
    display(tabela_aux_diagnostico_calibracao)

    limiar_completo = comp.get("limiar", LIMIARES_XGBOOST.get("completo_perinatal", 0.20))
    limiar_restritivo = rest.get("limiar", LIMIARES_XGBOOST.get("restritivo_final", 0.25))

    salvar_tabela(29, TABELAS[28][1], tabela_classificacao(comp["y_test"], comp["prob"], limiar=limiar_completo))
    salvar_tabela(30, TABELAS[29][1], tabela_matriz_confusao(comp["y_test"], comp["prob"], limiar=limiar_completo))

    # Tabelas 31 e 32: XGBoost para desfecho neonatal grave
    grave = treinar_xgb_cenario(data, features_restritivo, "DESFECHO_NEONATAL_GRAVE", "desfecho_neonatal_grave")
    salvar_tabela(31, TABELAS[30][1], pd.DataFrame([grave["metrics"]]))
    salvar_tabela(32, TABELAS[31][1], tabela_classificacao(grave["y_test"], grave["prob"], limiar=grave.get("limiar", 0.50)))

    # Tabela 33: concentração de risco por decil
    df_decil = data.loc[comp["idx_test"]].copy()
    df_decil["prob_xgb"] = comp["prob"]
    df_decil = df_decil.sort_values("prob_xgb", ascending=False)
    df_decil["decil_risco"] = pd.qcut(np.arange(len(df_decil)), 10, labels=[f"Decil {i}" for i in range(1,11)])
    tabela_33 = df_decil.groupby("decil_risco", observed=True)["NICU"].agg(total="count", casos_utin="sum", taxa_utin="mean").reset_index()
    tabela_33["taxa_utin_pct"] = tabela_33["taxa_utin"] * 100
    salvar_tabela(33, TABELAS[32][1], tabela_33)

    # Tabela 34: distribuição de UTIN e média de fatores por decil de risco na análise complementar
    fatores_top10 = [
        "PRETERM", "LOW_BW", "VERY_LOW_BW", "EXTREME_LOW_BW",
        "BAIXO_PESO", "MUITO_BAIXO_PESO", "EXTREMO_BAIXO_PESO",
        "GEMELAR", "PROM_PROXY_WIDE", "PROM_PROXY_CONSERV",
        "LD_CHOR", "LD_ANTB", "LD_STER", "RF_GDIAB_BIN", "RF_GHYPE_BIN", "WIC_BIN"
    ]
    fatores_top10 = [c for c in fatores_top10 if c in df_decil.columns]

    tabela_34 = df_decil.groupby("decil_risco", observed=True).agg(
        total=("NICU", "count"),
        casos_utin=("NICU", "sum"),
        taxa_utin=("NICU", "mean"),
        prob_media=("prob_xgb", "mean")
    ).reset_index()
    tabela_34["taxa_utin_pct"] = tabela_34["taxa_utin"] * 100

    for fator in fatores_top10:
        tabela_34[f"media_{fator}"] = (
            df_decil.groupby("decil_risco", observed=True)[fator]
            .mean()
            .reset_index(drop=True)
        )
        tabela_34[f"media_{fator}_pct"] = tabela_34[f"media_{fator}"] * 100

    salvar_tabela(34, TABELAS[33][1], tabela_34)

    # Tabela 35: principais fatores observados no Top 10% de maior risco
    n_top10 = max(int(np.ceil(len(df_decil) * 0.10)), 1)
    df_top10 = df_decil.head(n_top10).copy()

    linhas_top10 = []
    total_top10 = len(df_top10)
    for fator in fatores_top10:
        serie = pd.to_numeric(df_top10[fator], errors="coerce")
        positivos = int((serie == 1).sum())
        validos = int(serie.notna().sum())
        linhas_top10.append({
            "Fator": fator,
            "Registros válidos no Top 10%": validos,
            "Positivos no Top 10%": positivos,
            "Percentual no Top 10%": positivos / validos * 100 if validos else np.nan,
            "Total do Top 10%": total_top10
        })

    tabela_35 = (
        pd.DataFrame(linhas_top10)
        .sort_values("Percentual no Top 10%", ascending=False)
        .reset_index(drop=True)
    )
    tabela_35.insert(0, "Posição", range(1, len(tabela_35) + 1))
    salvar_tabela(35, TABELAS[34][1], tabela_35)

    # Tabela 36: concentração dos principais fatores no Top 10% de maior risco
    linhas_concentracao = []
    for fator in fatores_top10:
        serie_total = pd.to_numeric(df_decil[fator], errors="coerce")
        total_fator = int((serie_total == 1).sum())
        top10_fator = int((pd.to_numeric(df_top10[fator], errors="coerce") == 1).sum())
        linhas_concentracao.append({
            "Fator": fator,
            "Positivos na amostra de teste": total_fator,
            "Positivos no Top 10%": top10_fator,
            "Percentual dos positivos concentrados no Top 10%": top10_fator / total_fator * 100 if total_fator else np.nan,
            "Participação do Top 10% na amostra": len(df_top10) / len(df_decil) * 100 if len(df_decil) else np.nan
        })

    tabela_36 = (
        pd.DataFrame(linhas_concentracao)
        .sort_values("Percentual dos positivos concentrados no Top 10%", ascending=False)
        .reset_index(drop=True)
    )
    tabela_36.insert(0, "Posição", range(1, len(tabela_36) + 1))
    salvar_tabela(36, TABELAS[35][1], tabela_36)

    # Tabela 37: importância XGBoost
    tabela_37 = pd.DataFrame({"Variável": comp["features"], "Importância": comp["modelo_base"].feature_importances_}).sort_values("Importância", ascending=False)
    salvar_tabela(37, TABELAS[36][1], tabela_37)

    # Gráficos 14 a 19
    salvar_curva_roc(
        comp["y_test"],
        comp["prob"],
        GRAFICOS[13][1],
        "grafico_14_roc_xgb_completo_calibrado.png",
        prob_sem_calibracao=comp.get("prob_sem_calibracao"),
    )
    salvar_curva_pr(
        comp["y_test"],
        comp["prob"],
        GRAFICOS[14][1],
        "grafico_15_pr_xgb_completo_calibrado.png",
        prob_sem_calibracao=comp.get("prob_sem_calibracao"),
    )
    salvar_curva_roc(
        rest["y_test"],
        rest["prob"],
        GRAFICOS[15][1],
        "grafico_16_roc_xgb_restritivo_calibrado.png",
        prob_sem_calibracao=rest.get("prob_sem_calibracao"),
    )
    salvar_curva_pr(
        rest["y_test"],
        rest["prob"],
        GRAFICOS[16][1],
        "grafico_17_pr_xgb_restritivo_calibrado.png",
        prob_sem_calibracao=rest.get("prob_sem_calibracao"),
    )
    salvar_calibracao(
        comp["y_test"],
        comp["prob"],
        GRAFICOS[17][1],
        "grafico_18_calibracao_xgb_completo.png",
        prob_sem_calibracao=comp.get("prob_sem_calibracao"),
    )
    salvar_calibracao(
        rest["y_test"],
        rest["prob"],
        GRAFICOS[18][1],
        "grafico_19_calibracao_xgb_restritivo.png",
        prob_sem_calibracao=rest.get("prob_sem_calibracao"),
    )

    # Gráficos 20 e 21
    for res, num, arquivo in [(comp, 20, "grafico_20_importancia_completo.png"), (rest, 21, "grafico_21_importancia_restritivo.png")]:
        imp = pd.DataFrame({"Variável": res["features"], "Importância": res["modelo_base"].feature_importances_}).sort_values("Importância", ascending=False).head(15)
        fig, ax = plt.subplots(figsize=(8,6))
        ax.barh(
            imp["Variável"][::-1],
            imp["Importância"][::-1],
            label="XGBoost"
        )
        ax.set_title(GRAFICOS[num-1][1])
        ax.set_xlabel("Importância relativa")
        aplicar_legenda_padrao_notebook02(ax, loc="lower right")
        fig.tight_layout()
        fig.savefig(DIR_GRAFICOS / arquivo, dpi=300, bbox_inches="tight")
        plt.show()

    # Tabelas 38 e 39: SHAP robusto sem importar shap/numba
    # Prioridade:
    # 1) contribuições tipo SHAP nativas do XGBoost: Booster.predict(pred_contribs=True);
    # 2) fallback final para feature_importances_, identificado na tabela como não-SHAP.
    def calcular_shap_robusto_xgb(modelo, X_test, features, top_n=15, sample_size=5000):
        """
        Calcula importância média absoluta por contribuição Tree SHAP usando o XGBoost nativo.
        Não importa o pacote shap e não depende de Numba.

        Retorna:
            valores_importancia: np.ndarray com tamanho igual ao número de features.
            metodo: descrição do método usado.
        """
        features = list(features)
        sample = X_test.loc[:, features].copy()
        sample = sample.sample(min(sample_size, len(sample)), random_state=RANDOM_STATE)

        # Garante matriz numérica e preserva nomes de variáveis.
        sample = sample.apply(pd.to_numeric, errors="coerce").fillna(0.0)

        try:
            import xgboost as _xgb

            booster = modelo.get_booster()
            dmatrix = _xgb.DMatrix(sample, feature_names=features)

            # pred_contribs=True gera contribuições Tree SHAP.
            # A última coluna é o bias/base value e deve ser removida.
            contribs = booster.predict(dmatrix, pred_contribs=True)

            # Formatos possíveis:
            # - binário/regressão: (n_amostras, n_features + 1)
            # - multiclasse/strict: (n_amostras, n_classes, n_features + 1)
            # - variantes: (n_amostras, n_features + 1, n_classes)
            contribs = np.asarray(contribs)

            if contribs.ndim == 3:
                if contribs.shape[2] == len(features) + 1:
                    # (n, classes, features+1): seleciona a classe positiva quando existir.
                    classe = 1 if contribs.shape[1] > 1 else 0
                    contribs = contribs[:, classe, :]
                elif contribs.shape[1] == len(features) + 1:
                    # (n, features+1, classes): seleciona a classe positiva quando existir.
                    classe = 1 if contribs.shape[2] > 1 else 0
                    contribs = contribs[:, :, classe]
                else:
                    raise ValueError(f"Formato inesperado de contribuições tipo SHAP: {contribs.shape}")

            if contribs.ndim != 2:
                raise ValueError(f"Formato inesperado de contribuições tipo SHAP: {contribs.shape}")

            if contribs.shape[1] == len(features) + 1:
                contribs = contribs[:, :-1]  # remove bias/base value

            if contribs.shape[1] != len(features):
                raise ValueError(
                    f"Número de contribuições ({contribs.shape[1]}) diferente do número de features ({len(features)})."
                )

            valores_importancia = np.abs(contribs).mean(axis=0)
            metodo = "XGBoost pred_contribs=True — Tree SHAP nativo sem shap/numba"
            return valores_importancia, metodo

        except Exception as e:
            print("contribuições tipo SHAP nativas do XGBoost falhou; usando feature_importances_ como fallback:", e)

            valores_importancia = np.asarray(getattr(modelo, "feature_importances_", np.zeros(len(features))), dtype=float)
            if len(valores_importancia) != len(features):
                valores_importancia = np.zeros(len(features), dtype=float)

            metodo = "fallback feature_importances_ — não baseado em contribuições"
            return valores_importancia, metodo

    shap_abs, metodo_shap = calcular_shap_robusto_xgb(comp["modelo_base"], comp["X_test"], comp["features"])
    tabela_38 = (
        pd.DataFrame({"Variável": comp["features"], "Contribuição média absoluta": shap_abs})
        .sort_values("Contribuição média absoluta", ascending=False)
        .head(15)
        .reset_index(drop=True)
    )
    tabela_38.insert(0, "Posição", range(1, len(tabela_38) + 1))
    tabela_38["Método"] = metodo_shap
    salvar_tabela(38, TABELAS[37][1], tabela_38)

    total_shap = tabela_38["Contribuição média absoluta"].sum()
    tabela_39 = tabela_38.copy()
    tabela_39["Percentual"] = np.where(total_shap > 0, tabela_39["Contribuição média absoluta"] / total_shap * 100, 0)
    tabela_39["Percentual acumulado"] = tabela_39["Percentual"].cumsum()
    salvar_tabela(39, TABELAS[38][1], tabela_39)

    # Tabela 40: desempenho dos modelos seguros
    # Correção 0:
    # - valores ausentes são imputados com medianas calculadas apenas no treino;
    # - Regressão Logística, Random Forest e XGBoost recebem a mesma matriz imputada;
    # - isso evita erro por NaN e mantém o teste sem vazamento de informação.
    X_sec_raw, y_sec, feats_sec, _ = preparar_xy(data, features_restritivo, "NICU")
    Xtr_raw, Xte_raw, ytr, yte = train_test_split(
        X_sec_raw,
        y_sec,
        test_size=TEST_SIZE_XGBOOST_FINAL,
        stratify=y_sec,
        random_state=RANDOM_STATE
    )
    Xtr, Xte, medianas_seguras = imputar_por_mediana_treino(Xtr_raw, Xte_raw)

    pd.DataFrame({
        "variavel": medianas_seguras.index,
        "mediana_imputacao_calculada_no_treino": medianas_seguras.values
    }).to_csv(
        DIR_TABELAS / "tabela_aux_medianas_imputacao_modelos_seguros.csv",
        index=False,
        encoding="utf-8-sig"
    )

    modelos_seg = {
        "Regressão Logística": Pipeline([
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(max_iter=5000, class_weight="balanced"))
        ]),
        "Random Forest": RandomForestClassifier(
            n_estimators=300,
            max_depth=10,
            min_samples_leaf=10,
            class_weight="balanced",
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    }
    if XGBOOST_AVAILABLE:
        modelos_seg["XGBoost"] = criar_xgb(scale_pos_weight=(ytr == 0).sum() / max((ytr == 1).sum(), 1))

    linhas = []
    for nome, modelo in modelos_seg.items():
        modelo.fit(Xtr, ytr)
        pr = modelo.predict_proba(Xte)[:, 1]
        linha = avaliar_prob(yte, pr, nome, limiar=0.50)
        linha["observacao_metodologica"] = "Preditores restritivos imputados com medianas calculadas apenas no treino."
        linhas.append(linha)

    tabela_40 = pd.DataFrame(linhas)
    salvar_tabela(40, TABELAS[39][1], tabela_40)

    # Tabela 41 e 42: Random Forest
    rf = modelos_seg["Random Forest"]
    prob_rf = rf.predict_proba(Xte)[:,1]
    salvar_tabela(41, TABELAS[40][1], tabela_classificacao(yte, prob_rf, limiar=0.50))

    # Validação cruzada do Random Forest com controle de memória.
    # A base completa possui mais de 3,6 milhões de registros; repetir 5 ajustes completos
    # do Random Forest em paralelo pode estourar a memória em máquinas locais.
    # Por isso, a CV da Tabela 42 é feita em amostra estratificada reprodutível,
    # sem alterar o treino/teste principal nem os resultados finais do XGBoost.
    CV_SAMPLE_MAX = int(os.environ.get("TCC_RF_CV_SAMPLE_MAX", "200000"))
    if len(y_sec) > CV_SAMPLE_MAX:
        X_cv, _, y_cv, _ = train_test_split(
            X_sec_raw,
            y_sec,
            train_size=CV_SAMPLE_MAX,
            stratify=y_sec,
            random_state=RANDOM_STATE
        )
        obs_cv = f"CV estratificada em amostra reprodutível de n={len(y_cv):,} registros por controle de memória; imputação feita dentro de cada dobra."
    else:
        X_cv, y_cv = X_sec_raw, y_sec
        obs_cv = f"CV estratificada na base completa de n={len(y_cv):,} registros; imputação feita dentro de cada dobra."

    rf_cv = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", RandomForestClassifier(
            n_estimators=100,
            max_depth=10,
            min_samples_leaf=10,
            class_weight="balanced",
            random_state=RANDOM_STATE,
            n_jobs=1
        ))
    ])

    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
    auc_cv = cross_val_score(
        rf_cv,
        X_cv,
        y_cv,
        scoring="roc_auc",
        cv=cv,
        n_jobs=1,
        error_score="raise"
    )

    tabela_42 = pd.DataFrame([{
        "AUC_treino": roc_auc_score(ytr, rf.predict_proba(Xtr)[:,1]),
        "AUC_teste": roc_auc_score(yte, prob_rf),
        "AUC_CV_media": auc_cv.mean(),
        "AUC_CV_desvio": auc_cv.std(),
        "Observacao_CV": obs_cv
    }])
    salvar_tabela(42, TABELAS[41][1], tabela_42)

    # Tabela 43 a 45
    tabela_43 = pd.DataFrame([comp["metrics"], rest["metrics"]])
    salvar_tabela(43, TABELAS[42][1], tabela_43)
    salvar_tabela(44, TABELAS[43][1], tabela_matriz_confusao(comp["y_test"], comp["prob"], limiar=limiar_completo))
    salvar_tabela(45, TABELAS[44][1], tabela_matriz_confusao(rest["y_test"], rest["prob"], limiar=limiar_restritivo))

print("Modelagem, tabelas 29 a 45 e gráficos 14 a 21 concluídos quando dependências disponíveis.")
````

````python
def salvar_diagrama_graphviz(nome, titulo, edges):
    """
    Salva diagrama metodológico preferencialmente com Graphviz.
    Se Graphviz não estiver disponível, gera PNG com Matplotlib,
    evitando fallback apenas textual em Mermaid.
    """
    try:
        from graphviz import Digraph

        dot = Digraph(comment=titulo, format="png")
        dot.attr(rankdir="LR")
        for a, b in edges:
            dot.node(a)
            dot.node(b)
            dot.edge(a, b)

        caminho = DIR_FIGURAS / nome
        dot.render(str(caminho), cleanup=True)
        return caminho.with_suffix(".png")

    except Exception as e:
        # Fallback portátil: gera PNG com Matplotlib, sem depender de Graphviz.
        caminho = DIR_FIGURAS / f"{nome}.png"

        nodes = []
        for a, b in edges:
            if a not in nodes:
                nodes.append(a)
            if b not in nodes:
                nodes.append(b)

        # Posição simples em linha, suficiente para documentação do pipeline.
        x_positions = np.linspace(0, max(len(nodes) - 1, 1), len(nodes))
        y_positions = np.zeros(len(nodes))
        pos = {node: (x_positions[i], y_positions[i]) for i, node in enumerate(nodes)}

        largura = max(10, len(nodes) * 1.8)
        fig, ax = plt.subplots(figsize=(largura, 4.8))

        for node, (x, y) in pos.items():
            ax.text(
                x, y, node,
                ha="center", va="center",
                bbox=dict(boxstyle="round,pad=0.35", fc="white", ec="black", lw=1),
                fontsize=9,
                wrap=True
            )

        for a, b in edges:
            x1, y1 = pos[a]
            x2, y2 = pos[b]
            ax.annotate(
                "",
                xy=(x2 - 0.25, y2),
                xytext=(x1 + 0.25, y1),
                arrowprops=dict(arrowstyle="->", lw=1)
            )

        ax.set_title(titulo)
        ax.set_axis_off()
        ax.set_xlim(-0.7, max(x_positions) + 0.7)
        ax.set_ylim(-1, 1)
        fig.tight_layout()
        fig.savefig(caminho, dpi=300, bbox_inches="tight")
        plt.close(fig)

        print("Graphviz indisponível; diagrama salvo como PNG via Matplotlib:", caminho)
        return caminho


fig1_edges = [
    ("Natality Data 2024", "Carga do Parquet do notebook 01"),
    ("Carga do Parquet do notebook 01", "Parquet do notebook 01"),
    ("Parquet do notebook 01", "Base analítica"),
    ("Base analítica", "EDA"),
    ("Base analítica", "Regressão logística"),
    ("Base analítica", "Modelos supervisionados"),
    ("Modelos supervisionados", "Calibração e limiares"),
    ("Calibração e limiares", "Decis de risco"),
    ("Modelos supervisionados", "Contribuições tipo SHAP nativas e importância"),
    ("EDA", "Tabelas e gráficos finais"),
]

fig2_edges = [
    ("Problema de pesquisa", "Objetivos"),
    ("Objetivos", "Método retrospectivo"),
    ("Método retrospectivo", "Seleção de variáveis"),
    ("Seleção de variáveis", "Proxy RPM/PPROM"),
    ("Proxy RPM/PPROM", "Análise exploratória"),
    ("Análise exploratória", "Modelagem"),
    ("Modelagem", "Discussão crítica"),
    ("Discussão crítica", "Conclusões e limites"),
]

salvar_diagrama_graphviz("figura_01_pipeline_analitico", FIGURAS[0][1], fig1_edges)
salvar_diagrama_graphviz("figura_02_diagrama_metodologico", FIGURAS[1][1], fig2_edges)
````

````python
quadros_dict = {}

quadros_dict[4] = pd.DataFrame([
    [
        "PROM_PROXY_CONSERV",
        "PRETERM = 1 e LD_CHOR = 1",
        "Priorizar maior especificidade analítica",
        "Representa nascimentos pré-termo com registro de corioamnionite. Trata-se de definição mais restritiva, sem equivalência com diagnóstico clínico individual de RPM/PPROM."
    ],
    [
        "PROM_PROXY_WIDE",
        "PRETERM = 1 e (LD_CHOR = 1 ou LD_ANTB = 1 ou LD_STER = 1)",
        "Priorizar maior sensibilidade analítica para contexto de vulnerabilidade pré-termo",
        "Representa nascimentos pré-termo com corioamnionite, antibioticoterapia intraparto ou uso de corticoide antenatal/esteroides. É uma definição ampliada, podendo incluir diferentes contextos obstétricos e assistenciais não exclusivos de RPM/PPROM; portanto, não deve ser interpretada como confirmação clínica nem como aproximação diagnóstica individual."
    ],
], columns=["Proxy operacional", "Regra lógica operacional", "Finalidade analítica", "Interpretação metodológica"])

quadros_dict[6] = quadros_dict[4].copy()

quadros_dict[9] = pd.DataFrame([
    ["Registros brutos/intermediários", len(df)],
    ["Registros na base analítica", len(data)],
    ["Colunas na base analítica", data.shape[1]],
    ["Registros com desfecho NICU válido", data["NICU"].notna().sum()],
], columns=["Etapa", "Quantidade"])

quadros_dict[12] = pd.DataFrame([
    ["Completo/perinatal", ", ".join(features_completo), "Máximo desempenho retrospectivo em registros consolidados; maior risco de vazamento temporal por variáveis neonatais próximas do desfecho"],
    ["Restritivo final", ", ".join(features_restritivo), "Menor risco de vazamento temporal; mantém COMBGEST e DBWT, portanto reduz mas não elimina a proximidade temporal com o desfecho"],
], columns=["Cenário", "Variáveis", "Finalidade"])

quadros_dict[15] = pd.DataFrame([
    ["AUC-ROC", "Discriminação global"],
    ["PR-AUC", "Desempenho em contexto de desbalanceamento"],
    ["Brier Score", "Qualidade probabilística/calibração"],
    ["Precisão", "Proporção de positivos previstos corretamente"],
    ["Sensibilidade", "Capacidade de identificar positivos reais"],
    ["F1-score", "Síntese entre precisão e sensibilidade"],
], columns=["Métrica", "Interpretação"])

quadros_dict[23] = tabela_3.copy()
quadros_dict[24] = tabela_4.copy()
quadros_dict[29] = pd.DataFrame([
    ["Prematuridade e baixo peso", "Concentram vulnerabilidade neonatal e maior frequência observada de UTIN"],
    ["Gemelaridade e pluralidade", "Associam-se a maior risco obstétrico e neonatal em registros populacionais"],
    ["Proxy operacional ampla", "Funciona como marcador operacional ampliado de vulnerabilidade pré-termo, sem equivalência diagnóstica individual de RPM/PPROM"],
    ["Decil de maior risco", "Organiza retrospectivamente registros com maior concentração de fatores de vulnerabilidade"],
], columns=["Achado complementar", "Diálogo com a literatura médica"])

quadros_dict[30] = pd.DataFrame({"Variável removida do cenário restritivo": variaveis_removidas_vazamento})

quadros_dict[42] = pd.DataFrame([
    ["Ingestão de dados", "Carga do Parquet produzido pelo notebook 01"],
    ["Curadoria", "Limpeza, recodificação e tratamento de ausentes"],
    ["Engenharia", "Criação de variáveis derivadas e proxies"],
    ["EDA", "Tabelas, gráficos e análise descritiva"],
    ["Modelagem", "Regressão logística, Random Forest e XGBoost"],
    ["Validação", "Calibração, limiares, matriz de confusão e decis"],
    ["Explicabilidade", "Importância de variáveis e contribuições tipo SHAP nativas do XGBoost"],
    ["Exportação", "Tabelas, gráficos, quadros e manifesto"],
], columns=["Módulo", "Descrição"])

# Quadros documentais genéricos para todos os demais itens.
for numero, titulo, tipo in QUADROS:
    if numero not in quadros_dict:
        quadros_dict[numero] = pd.DataFrame([
            ["Finalidade", titulo],
            ["Natureza", tipo],
            ["Status no notebook", "Gerado como quadro documental editável"],
            ["Observação", "Revisar redação final antes de inserir no relatório"]
        ], columns=["Campo", "Descrição"])

for numero, titulo, tipo in QUADROS:
    salvar_quadro(numero, titulo, quadros_dict[numero])

print(f"{len(quadros_dict)} quadros exportados.")
````

## 13. Relatório automático de conferência: item do TCC × arquivo gerado

O relatório de conferência verifica se cada quadro, figura, gráfico e tabela previsto no manifesto foi efetivamente gerado ou documentado.

Essa checagem é importante para a entrega final porque identifica pendências, evita ausência de itens citados no texto e reforça que o relatório foi construído a partir das saídas do notebook.

Nesta versão, o manifesto também reconhece os **Quadros A1 e A2** do apêndice, preservando a numeração alfanumérica usada no documento final.

````python
def existe_arquivo_com_prefixo(pasta, prefixo):
    return any(p.name.startswith(prefixo) for p in pasta.glob("*"))

manifesto = []

for numero, titulo, tipo in QUADROS:
    prefixo = f"quadro_{numero:02d}_"
    ok = existe_arquivo_com_prefixo(DIR_QUADROS, prefixo)
    manifesto.append(["Quadro", numero, titulo, "OK" if ok else "PENDENTE", str(DIR_QUADROS), "documental/computacional"])

# Quadros de apêndice presentes no documento final.
# Eles entram no manifesto com numeração alfanumérica para preservar o título oficial do relatório.
for numero, titulo, tipo in globals().get("QUADROS_APENDICE", []):
    prefixo = f"quadro_{str(numero).lower()}_"
    ok = existe_arquivo_com_prefixo(DIR_QUADROS, prefixo)
    manifesto.append(["Quadro", numero, titulo, "OK" if ok else "DOCUMENTAL", str(DIR_QUADROS), "apêndice/documental"])

for numero, titulo in FIGURAS:
    prefixo = f"figura_{numero:02d}_"
    ok = existe_arquivo_com_prefixo(DIR_FIGURAS, prefixo)
    # Nomes foram salvos como figura_01_pipeline_analitico e figura_02_diagrama_metodologico
    ok = ok or existe_arquivo_com_prefixo(DIR_FIGURAS, f"figura_0{numero}_")
    manifesto.append(["Figura", numero, titulo, "OK" if ok else "PENDENTE", str(DIR_FIGURAS), "diagrama"])

for numero, titulo in GRAFICOS:
    prefixo = f"grafico_{numero:02d}_"
    ok = existe_arquivo_com_prefixo(DIR_GRAFICOS, prefixo)
    manifesto.append(["Gráfico", numero, titulo, "OK" if ok else "PENDENTE", str(DIR_GRAFICOS), "imagem"])

for numero, titulo in TABELAS:
    prefixo = f"tabela_{numero:02d}_"
    ok = existe_arquivo_com_prefixo(DIR_TABELAS, prefixo)
    manifesto.append(["Tabela", numero, titulo, "OK" if ok else "PENDENTE", str(DIR_TABELAS), "csv"])

manifesto_df = pd.DataFrame(manifesto, columns=["Tipo", "Número", "Título", "Status", "Pasta", "Observação"])
manifesto_df.to_csv(DIR_MANIFESTO / "manifesto_tcc_item_por_item.csv", index=False, encoding="utf-8-sig")

resumo_manifesto = manifesto_df.groupby(["Tipo", "Status"]).size().reset_index(name="quantidade")
resumo_manifesto.to_csv(DIR_MANIFESTO / "resumo_manifesto_tcc.csv", index=False, encoding="utf-8-sig")

display(Markdown("### Manifesto item por item — TCC"))
display(manifesto_df)

display(Markdown("### Resumo do manifesto — TCC"))
display(resumo_manifesto)
````

````python
# Checklist final de reprodutibilidade
# Deve ser executado após a geração das tabelas, gráficos, quadros, figuras e manifesto.
# A contagem oficial vem do manifesto, não do número de arquivos nas pastas, para evitar contar
# arquivos auxiliares, versões antigas ou duplicatas.

def _contagem_manifesto(tipo, status="OK"):
    if "manifesto_df" not in globals():
        return 0
    m = manifesto_df
    return int(((m["Tipo"] == tipo) & (m["Status"] == status)).sum())

n_base_carregada = int(len(df)) if "df" in globals() else None
n_base_derivada = int(len(data)) if "data" in globals() else None
if "data" in globals():
    _vars_chave_checklist = globals().get(
        "VARIAVEIS_CHAVE_AMOSTRA_ANALITICA",
        ["NICU", "COMBGEST", "DBWT", "DPLURAL", "GEMELAR", "PRETERM", "LOW_BW", "PROM_PROXY_WIDE", "PROM_PROXY_CONSERV"]
    )
    _vars_chave_checklist = [c for c in _vars_chave_checklist if c in data.columns]
    n_amostra_analitica_final = int(data.dropna(subset=_vars_chave_checklist).shape[0]) if _vars_chave_checklist else None
else:
    n_amostra_analitica_final = None

n_excluidos_incompletude = (
    n_base_derivada - n_amostra_analitica_final
    if n_base_derivada is not None and n_amostra_analitica_final is not None
    else None
)
pct_excluidos_incompletude = (
    n_excluidos_incompletude / n_base_derivada * 100
    if n_base_derivada not in (None, 0) and n_excluidos_incompletude is not None
    else None
)

checklist_reprodutibilidade = pd.DataFrame([
    ["Versão do pipeline registrada", "OK" if "VERSAO_PIPELINE" in globals() else "PENDENTE", globals().get("VERSAO_PIPELINE", None)],
    ["Ambiente registrado", "OK" if "VERSOES_AMBIENTE" in globals() else "PENDENTE", len(VERSOES_AMBIENTE) if "VERSOES_AMBIENTE" in globals() else 0],
    ["Configuração de aceleração XGBoost registrada", "OK" if "PARAMETROS_XGBOOST_ACELERACAO" in globals() else "PENDENTE", json.dumps(PARAMETROS_XGBOOST_ACELERACAO, ensure_ascii=False) if "PARAMETROS_XGBOOST_ACELERACAO" in globals() else None],
    ["Limiares XGBoost registrados", "OK" if "LIMIARES_XGBOOST" in globals() else "PENDENTE", json.dumps(LIMIARES_XGBOOST, ensure_ascii=False) if "LIMIARES_XGBOOST" in globals() else None],
    ["Partição de teste XGBoost final", "OK" if "TEST_SIZE_XGBOOST_FINAL" in globals() else "PENDENTE", TEST_SIZE_XGBOOST_FINAL if "TEST_SIZE_XGBOOST_FINAL" in globals() else None],
    ["Base carregada/processada", "OK" if n_base_carregada and n_base_carregada > 0 else "PENDENTE", n_base_carregada],
    ["Base derivada com variáveis criadas", "OK" if n_base_derivada and n_base_derivada > 0 else "PENDENTE", n_base_derivada],
    ["Amostra analítica final nas variáveis-chave", "OK" if n_amostra_analitica_final and n_amostra_analitica_final > 0 else "PENDENTE", n_amostra_analitica_final],
    ["Registros excluídos por incompletude nas variáveis-chave", "OK" if n_excluidos_incompletude is not None else "PENDENTE", n_excluidos_incompletude],
    ["Percentual excluído por incompletude nas variáveis-chave (%)", "OK" if pct_excluidos_incompletude is not None else "PENDENTE", pct_excluidos_incompletude],
    ["Proxy ampla criada", "OK" if "data" in globals() and "PROM_PROXY_WIDE" in data.columns else "PENDENTE", None],
    ["Proxy conservadora criada", "OK" if "data" in globals() and "PROM_PROXY_CONSERV" in data.columns else "PENDENTE", None],
    ["Tabelas oficiais no manifesto", "OK" if _contagem_manifesto("Tabela") == len(TABELAS) else "PENDENTE", _contagem_manifesto("Tabela")],
    ["Gráficos oficiais no manifesto", "OK" if _contagem_manifesto("Gráfico") == len(GRAFICOS) else "PENDENTE", _contagem_manifesto("Gráfico")],
    ["Quadros oficiais no manifesto", "OK" if _contagem_manifesto("Quadro") == len(QUADROS) else "PENDENTE", _contagem_manifesto("Quadro")],
    ["Figuras oficiais no manifesto", "OK" if _contagem_manifesto("Figura") == len(FIGURAS) else "PENDENTE", _contagem_manifesto("Figura")],
    ["Manifesto gerado", "OK" if "manifesto_df" in globals() and len(manifesto_df) > 0 else "PENDENTE", len(manifesto_df) if "manifesto_df" in globals() else 0],
    ["Excel consolidado", "OK" if "arquivo_excel" in globals() and Path(arquivo_excel).exists() else "PENDENTE", str(arquivo_excel) if "arquivo_excel" in globals() else None],
], columns=["Item", "Status", "Valor/Detalhe"])

checklist_reprodutibilidade.to_csv(
    DIR_MANIFESTO / "checklist_reprodutibilidade.csv",
    index=False,
    encoding="utf-8-sig"
)

ambiente_df = pd.DataFrame(
    [{"Componente": k, "Versão": v} for k, v in VERSOES_AMBIENTE.items()]
)
ambiente_df.to_csv(
    DIR_MANIFESTO / "ambiente_execucao.csv",
    index=False,
    encoding="utf-8-sig"
)

with open(DIR_MANIFESTO / "ambiente_execucao.txt", "w", encoding="utf-8") as f:
    f.write("Ambiente de execução do pipeline reprodutível do TCC\n")
    f.write("=================================================\n\n")
    for k, v in VERSOES_AMBIENTE.items():
        f.write(f"{k}: {v}\n")

with open(DIR_MANIFESTO / "requirements_pipeline_tcc.txt", "w", encoding="utf-8") as f:
    for pacote in ["numpy", "pandas", "matplotlib", "scikit-learn", "xgboost", "statsmodels", "openpyxl"]:
        versao = VERSOES_AMBIENTE.get(pacote, "")
        f.write(f"{pacote}=={versao}\n" if versao else f"{pacote}\n")

display(Markdown("### Checklist final de reprodutibilidade"))
display(checklist_reprodutibilidade)

display(Markdown("### Ambiente de execução"))
display(ambiente_df)

print("Checklist, ambiente e requirements salvos em:", DIR_MANIFESTO)
````

## 15. Limitações, uso responsável e orientação de execução

A interpretação dos resultados deve respeitar os limites metodológicos do estudo:

- a pesquisa é retrospectiva e observacional;
- a base utilizada contém registros secundários públicos e consolidados;
- a proxy operacional de RPM/PPROM não é diagnóstico individual, não confirma RPM/PPROM e não estima a prevalência clínica real dessa condição;
- a `PROM_PROXY_WIDE` deve ser compreendida como marcador operacional ampliado de vulnerabilidade pré-termo, pois inclui indicadores que também podem refletir diferentes contextos obstétricos e assistenciais;
- na definição ampla, registros com `PRETERM` informado e sem componentes positivos são classificados como negativos para a proxy; essa decisão preserva a análise binária, mas pode subestimar registros potencialmente compatíveis quando houver ausência ou sub-registro dos componentes;
- a internação em UTIN deve ser interpretada como desfecho operacional, influenciado por gravidade neonatal, protocolos assistenciais e disponibilidade de recursos;
- o modelo completo/perinatal pode incorporar variáveis temporalmente próximas do desfecho, como Apgar, ventilação, surfactante, antibiótico neonatal e convulsões;
- o cenário restritivo reduz o risco de vazamento temporal, mas ainda mantém idade gestacional e peso ao nascer; por isso, não deve ser interpretado como modelo preditivo pré-natal;
- as medidas de importância das variáveis e as contribuições tipo SHAP nativas do XGBoost representam o classificador base, enquanto as métricas probabilísticas finais derivam dos modelos calibrados;
- nenhum resultado deste notebook deve ser utilizado como ferramenta clínica prospectiva sem validação independente, avaliação de equidade, calibração em população-alvo e integração segura ao fluxo assistencial.

Em síntese, os modelos desenvolvidos neste estudo devem ser interpretados como instrumentos retrospectivos de análise e estratificação de risco em registros populacionais consolidados, e não como ferramentas clínicas prospectivas para tomada de decisão individual.

### Orientação de execução para entrega

Execute este notebook a partir da pasta do projeto, mantendo o arquivo `.txt` original ou o arquivo `.parquet` intermediário em uma das pastas previstas pelo pipeline: `dados_brutos`, `dados_intermediarios`, `dados`, `data`, `raw`, pasta atual ou `/mnt/data`.

Para uma versão final de entrega, recomenda-se executar o pipeline completo, conferir o manifesto, validar os arquivos gerados e, se necessário, limpar saídas locais antes do envio para preservar a portabilidade.

### Uso de GPU

Por padrão, o notebook usa `TCC_XGBOOST_DEVICE=auto`: ele tenta usar GPU CUDA quando `nvidia-smi` está disponível e usa CPU automaticamente caso contrário.

Para forçar CPU antes de abrir ou executar o notebook:

```bash
export TCC_XGBOOST_DEVICE=cpu
```

Para solicitar GPU explicitamente:

```bash
export TCC_XGBOOST_DEVICE=gpu
```

Em Google Colab, selecione um ambiente com GPU antes de executar o notebook. Se CUDA não estiver disponível, o pipeline continua reprodutível em CPU.

### Ordem recomendada para reprodução

1. Executar previamente o notebook 01 e disponibilizar `natality_data_2024.parquet` em `data/processed/` ou em uma das pastas verificadas pelo notebook 02.
2. Colocar o arquivo em `dados_brutos/`, `dados_intermediarios/`, `dados/`, `data/`, `raw/` ou na mesma pasta do notebook.
3. Executar o notebook inteiro, do início ao fim.
4. Conferir o manifesto em `outputs/manifesto/` e o arquivo consolidado `outputs/resultados_tcc_pipeline_completo.xlsx`.
5. Atualizar o relatório textual somente com tabelas, quadros, figuras e gráficos marcados como `OK` no manifesto.

### Redação defensiva recomendada para o relatório

Use a formulação: **“proxy operacional compatível com contexto de vulnerabilidade pré-termo associada a marcadores obstétricos/assistenciais disponíveis na base, sem equivalência diagnóstica individual com RPM/PPROM”**.

Evite afirmar que a proxy mede RPM/PPROM real ou prevalência clínica da condição.

Os arquivos auxiliares não substituem as tabelas oficiais do TCC; eles servem como trilha de auditoria metodológica e reprodutibilidade.
